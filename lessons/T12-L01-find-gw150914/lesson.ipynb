```
# Setup: everything the lesson needs, in one cell, with versions printed.
import contextlib
import io
import sys
import time
import traceback
from pathlib import Path
from typing import Callable

import numpy as np

_LESSON_T0 = time.perf_counter()
print("python", sys.version.split()[0], "· numpy", np.__version__, "· platform", sys.platform)

_FAILED_CHECKS: list[str] = []
_STATUS: dict[str, str] = {}   # label -> "passed" | "failed" | "not started", latest run

# The exercises, in the order you meet them, and the functions each one asks you to write.
_EXERCISES: dict[str, tuple[str, ...]] = {
    "exercise 1": ("welch_psd",),
    "exercise 2": ("whiten",),
    "exercise 3": ("bandpass_fd",),
    "exercise 4": ("chirp_frequency",),
    "exercise 5": ("chirp_phase",),
    "exercise 6": ("matched_filter_snr",),
    "exercise 7": ("estimate_chirp_mass",),
    "exercise 8": ("check_light_travel_consistency",),
}


def _named(labels: list[str]) -> str:
    if len(labels) == 1:
        return f"{labels[0]} ({', '.join(_EXERCISES[labels[0]])})"
    nums = [label.split()[-1] for label in labels]
    return "exercises " + ", ".join(nums[:-1]) + " and " + nums[-1]


# --- COMMONS TIME LIMIT v4 --- every check in this notebook runs under a time limit
import contextlib, gc, os, signal, subprocess, sys, threading, time


# A time limit for every check that calls the student's code: a loop that never ends must fail
# its check, not freeze the notebook or the grader (which imports the notebook). Copy this block
# unchanged; change only the three limits below it. Why it is this careful:
#  * Whoever runs the notebook may already have an alarm running (tools/grade.py stops every
#    rubric row after 45 s; a harness may have one whose default action ends the process). This
#    guard saves it, lets whichever deadline comes first win (handing the signal to the outer
#    handler when it is the outer's turn) and on the way out re-arms it with the time it had left
#    and its interval. Switching it off instead would let the NEXT hang run for ever.
#  * Its own alarm is one-shot, re-armed by hand every 0.25 s once the time is up (in case a bare
#    `except:` swallows the first). A second alarm alone is not enough: `while True: try: ...
#    except: pass` is interrupted inside the `try` every time and carries on. So from the second
#    stop on, a line tracer (sys.settrace, on the running frames too) raises the stop again at the
#    first line the code runs afterwards, which is in the handler, outside the `try`; it is put
#    back as it was on the way out. A repeating alarm kept firing while a runaway loop's huge list
#    was being torn down; a repeat that arrived after the default handler was back ended the
#    process (exit -14, a dead Jupyter kernel), and one that arrived inside contextlib's exit
#    escaped as a raw timeout and left this handler installed.
#    On Python 3.12+ that tracer is sys.monitoring, which stays on after it raises, so a loop
#    inside TWO bare `except:` blocks is stopped as well: the Nth stop lets it raise N-1 times in
#    a row (one per handler it has to leave), and no more, so the `finally:` blocks and context
#    managers that run as the stop travels out are not interrupted again. On 3.11 sys.settrace is
#    unset by the interpreter each time it raises, so only the innermost one is peeled per alarm.
#  * It also watches memory: every 0.1 s it compares the process's resident memory with what it
#    was when the block started, and stops the block as soon as the growth passes `memory_mib`.
#    A loop that appends to a list reaches gigabytes in seconds, long before any sensible time
#    limit, so the time limit alone does not bound it. (Resident memory NOW, not the peak so
#    far: a peak only ever rises, so after one runaway check the next would be allowed another
#    `memory_mib` on top of it, and five looping exercises would reach ten gigabytes. Where the
#    system will not say, /proc on Linux and libproc on macOS being the two that do, it falls
#    back to the peak.)
#  * It never raises inside its own set-up, exit or clean-up (nor inside another guard's: guards
#    nest, the grader's wraps this one), and on the way out it takes any alarm of its own that has
#    gone off but not yet been handled (with SIGALRM blocked) before the old handler is put back.
# tools/verify_all.py fails a lesson whose guard does less.
class _OutOfTime(BaseException):
    """Raised inside the guarded block when its time is up. A BaseException, so `except
    Exception:` in the code under test cannot swallow it; it repeats every 0.25 s, and from the
    second time on a line tracer raises it again at the next line the code runs (inside the
    handler that swallowed the first), in case a bare `except:` does. `_time_limit` turns it into
    an AssertionError once it has left that code."""


def _resident() -> float:
    """This process's resident memory now, in MiB (Linux: /proc; macOS: libproc); its peak so far elsewhere."""
    try:
        if sys.platform.startswith("linux"):
            with open("/proc/self/statm") as f:
                return int(f.read().split()[1]) * os.sysconf("SC_PAGE_SIZE") / 2**20
        if sys.platform == "darwin":
            import ctypes, struct
            info = ctypes.create_string_buffer(96)           # struct proc_taskinfo
            if ctypes.CDLL("/usr/lib/libproc.dylib").proc_pidinfo(os.getpid(), 4, 0, info, 96) == 96:
                return struct.unpack_from("QQ", info.raw)[1] / 2**20     # pti_resident_size
    except Exception:
        pass
    import resource
    return resource.getrusage(resource.RUSAGE_SELF).ru_maxrss / (2**20 if sys.platform == "darwin" else 2**10)


@contextlib.contextmanager
def _time_limit(seconds: float, what: str, memory_mib: float = 0):
    """Fail the block with an AssertionError naming `what` if it runs longer than `seconds`, or
    (when `memory_mib` is set) if the process's resident memory grows by more than that while it runs.
    Uses the operating system's alarm (Linux and macOS, main thread); a no-op elsewhere."""
    if not (all(hasattr(signal, name) for name in ("setitimer", "SIGALRM", "pthread_sigmask", "sigpending", "sigwait"))
            and threading.current_thread() is threading.main_thread()):
        yield
        return
    if getattr(_time_limit, "stopped", False):           # a stopped check's frames keep what they built in a
        _time_limit.stopped = False                      # reference cycle until now: free it, so it is not counted
        gc.collect()                                     # as the growth of this block
    message = (f"{what} was still running after {seconds:g} s: a loop whose stopping condition "
               "is never met runs for ever; give every loop a cap")
    started = time.monotonic()
    used_mib = None
    if memory_mib:
        used_mib = _resident
        base_mib = used_mib()                            # resident memory now; only growth beyond it counts
    previous = signal.getsignal(signal.SIGALRM)
    outer_left, outer_every = signal.getitimer(signal.ITIMER_REAL)   # an alarm already running
    # One alarm at a time, never a repeating one. next: when ours is due; outer: when the alarm we
    # found is due (every: its interval); unseen: an alarm is armed and has not reached _fire yet.
    state = {"next": started + seconds, "outer": started + outer_left if outer_left > 0 else None,
             "every": outer_every, "unseen": False, "fired": False, "done": False, "message": message, "count": 0, "budget": 0,
             "traced": False, "tracer": None, "tool": None, "at": None, "thread": threading.get_ident(),
             "me": sys._getframe(), "owner": None}
    owner = state["me"].f_back
    while owner is not None and owner.f_globals is vars(contextlib):    # contextlib's __enter__
        owner = owner.f_back
    state["owner"], state["at"] = owner, owner.f_lineno   # the frame running the `with`, and its line

    def _arm(now, soonest=0.001):                        # one shot, at whichever deadline is next
        due = [t for t in (state["next"], state["outer"]) if t is not None]
        if used_mib and state["next"] is not None and not state["fired"]:
            due.append(now + 0.1)                        # and look at the memory every 0.1 s
        if due:
            signal.setitimer(signal.ITIMER_REAL, max(min(due) - now, soonest))
            state["unseen"] = True

    def _in_guard(frame):                                # is the alarm landing in a guard's own set-up, exit
        while frame is not None and frame is not state["owner"]:     # or clean-up (this one's or another's:
            code = frame.f_code                          # guards nest), or in contextlib's __exit__ under
            if (frame is state["me"] or code in _own_code or code.co_name in _GUARDS   # a `with`?
                    or frame.f_globals.get("__name__") == "contextlib"):
                return True
            frame = frame.f_back
        return False

    def _fire(signum, frame):
        state["unseen"] = False
        if state["done"]:
            return                                       # a late alarm of ours: already handled
        if _in_guard(frame):                             # never raise inside the guard: it is
            _arm(time.monotonic(), 0.01)                 # finishing; look again shortly
            return
        now = time.monotonic()
        if state["outer"] is not None and now >= state["outer"] - 0.005:
            every = state["every"]                       # the outer alarm is due: act as it would
            state["outer"] = state["outer"] + every * (1 + (now - state["outer"]) // every) if every else None
            try:
                if callable(previous):
                    previous(signum, frame)
                elif previous != signal.SIG_IGN:         # its default action ends the process
                    signal.signal(signal.SIGALRM, signal.SIG_DFL)
                    os.kill(os.getpid(), signal.SIGALRM)
            except BaseException:
                state["next"] = None                     # the outer one is stopping the block
                raise
            finally:
                left, every = signal.getitimer(signal.ITIMER_REAL)
                if left > 0:                             # the outer handler armed its next alarm
                    state["outer"], state["every"] = time.monotonic() + left, every
                _arm(time.monotonic())
            now = time.monotonic()
        grown = used_mib() - base_mib if used_mib and not state["fired"] else 0
        if grown > memory_mib and state["next"] is not None:       # out of memory before out of time
            state["message"] = (f"{what} was using over {memory_mib:g} MiB more memory than when it started, "
                                f"after {now - started:.1f} s: a loop that keeps adding to a list or a string "
                                "never stops; give every loop a cap")
            state["next"] = now
        if state["next"] is not None and now >= state["next"] - 0.005:
            swallowed = state["fired"]                   # a second stop: the first did not end the block
            state["fired"] = True
            state["count"] += 1
            state["next"] = now + 0.25                   # again, in case a bare `except:` swallows it
            _arm(now)
            if swallowed:
                state["budget"] = state["count"] - 1     # the tracer raises this many times, then leaves clean-up alone
                _trace_from(frame)
            raise _OutOfTime(state["message"], state)
        _arm(now)

    def _trace(frame, event, arg):                       # raises the stop at the next line of code that
        if state["done"] or _in_guard(frame):            # runs: inside the handler that swallowed the last
            return None                                  # (the interpreter then unsets it; the next alarm
        if event == "line":                              # sets it again)
            raise _OutOfTime(state["message"], state)
        return _trace

    def _line(code, line):                               # Python 3.12+: sys.monitoring, which is not unset
        if state["done"] or state["budget"] <= 0 or threading.get_ident() != state["thread"]:   # not unset when a
            return                                       # callback raises: the next line the code runs raises too, handlers
        frame = sys._getframe(1)                         # included (this thread only), but only `budget` times:
        if _in_guard(frame) or (frame is state["owner"] and line == state["at"]):   # clean-up that runs once the stop has
            return                                       # left the swallowing loops must not be interrupted again
        state["budget"] -= 1                             # (not the `with` line: its exit runs the guard)
        raise _OutOfTime(state["message"], state)

    def _trace_from(frame):                              # trace the running frames, and the ones called next
        if not state["traced"]:
            state["traced"] = True
            mon = getattr(sys, "monitoring", None)
            for tool in (5, 4, 3, 2, 1, 0) if mon is not None else ():
                try:
                    mon.use_tool_id(tool, "commons-time-limit")
                except ValueError:                       # a debugger or a profiler has this one
                    continue
                state["tool"] = tool
                mon.register_callback(tool, mon.events.LINE, _line)
                mon.set_events(tool, mon.events.LINE)
                return
            state["tracer"] = sys.gettrace()
            sys.settrace(_trace)
        while frame is not None and frame is not state["owner"]:
            frame.f_trace = _trace
            frame = frame.f_back

    _GUARDS = ("_time_limit", "_deadline", "_resident", "_fire", "_arm", "_trace", "_trace_from", "_line", "_in_guard")
    _own_code = (_fire.__code__, _arm.__code__, _trace.__code__, _trace_from.__code__, _line.__code__)
    signal.signal(signal.SIGALRM, _fire)
    _arm(time.monotonic())
    try:
        try:
            yield
        finally:
            state["done"] = True                         # from here on an alarm of ours only counts
            if state["traced"]:
                state["traced"] = False
                if state["tool"] is not None:
                    sys.monitoring.set_events(state["tool"], 0)
                    sys.monitoring.register_callback(state["tool"], sys.monitoring.events.LINE, None)
                    sys.monitoring.free_tool_id(state["tool"])
                else:
                    sys.settrace(state["tracer"])        # the tracer that was there before, if any
            mask = signal.pthread_sigmask(signal.SIG_BLOCK, {signal.SIGALRM})
            try:
                if signal.setitimer(signal.ITIMER_REAL, 0)[0] == 0 and state["unseen"]:
                    # Ours went off and has not reached _fire: it is pending here, or on its way to
                    # another thread. Take it before the old handler (by default, one that ends the
                    # process) is back.
                    give_up = time.monotonic() + 0.5
                    while state["unseen"] and time.monotonic() < give_up:
                        if signal.SIGALRM in signal.sigpending():
                            signal.sigwait({signal.SIGALRM})
                            state["unseen"] = False
                        else:
                            time.sleep(0.001)            # another thread has it: _fire runs next
                signal.signal(signal.SIGALRM, signal.SIG_DFL if previous is None else previous)
                outer = state["outer"]                   # re-arm the alarm we found, last
                signal.setitimer(signal.ITIMER_REAL, *((max(outer - time.monotonic(), 0.001), state["every"])
                                                       if outer is not None else (0, 0)))
            finally:
                signal.pthread_sigmask(signal.SIG_SETMASK, mask)
                state["me"] = state["owner"] = None
    except _OutOfTime as exc:
        if exc.args[-1] is not state:                    # an enclosing guard's: not ours to report
            raise
        _time_limit.stopped = True
        raise AssertionError(state["message"]) from None
    if state["fired"]:                                   # the block caught our alarm and carried on
        _time_limit.stopped = True
        raise AssertionError(state["message"])


# The three limits. A correct answer needs a small fraction of the first two (at least five times
# less, measured on the worked solution, on the slower of a Mac and an emulated Linux machine), so
# only a loop that never stops, or one that keeps appending until memory runs out, ever meets one.
_CHECK_SECONDS = 2     # an exercise's check
_DEMO_SECONDS = 10     # any other cell that runs your code: a demo, a table, a figure
_MEMORY_MIB = 2048     # how far a check may grow the process's resident memory before it is stopped


def _limited(label: str, check):
    """Run `check()` under `_time_limit` and return what it returns: _CHECK_SECONDS for an
    exercise's check, _DEMO_SECONDS for anything else. A check that is stopped (too long, or too
    much memory) fails with an AssertionError that says so; whatever the stopped code had built (a
    runaway list can hold gigabytes) is freed at once, not whenever the garbage collector next
    runs; and every program it had started that is still running is stopped too."""
    seconds = _CHECK_SECONDS if label.lower().startswith("exercise") else _DEMO_SECONDS
    before = _processes()
    try:
        with _time_limit(seconds, "this check", _MEMORY_MIB):
            return check()
    except AssertionError as exc:
        if not str(exc).startswith("this check was "):
            raise                                        # the check's own verdict, unchanged
        message = str(exc)
    gc.collect()          # the stopped frames hold what they built in a reference cycle until now
    _stop_strays(before)
    raise AssertionError(message)


def _processes() -> dict:
    """{pid: (parent, process group)} of every process (Linux: /proc; macOS: ps); {} if unreadable."""
    table = {}
    try:
        if os.path.isdir("/proc/self"):
            for entry in os.listdir("/proc"):
                if entry.isdigit():
                    try:
                        with open(f"/proc/{entry}/stat") as f:
                            fields = f.read().rsplit(")", 1)[1].split()
                        table[int(entry)] = (int(fields[1]), int(fields[2]))
                    except (OSError, ValueError, IndexError):
                        continue
        else:
            listing = subprocess.run(["ps", "-A", "-o", "pid=,ppid=,pgid="], capture_output=True, text=True,
                                     timeout=10).stdout
            for row in listing.splitlines():
                pid, ppid, pgid = (int(x) for x in row.split())
                table[pid] = (ppid, pgid)
    except (OSError, ValueError, subprocess.SubprocessError):
        return {}
    return table


def _stop_strays(before: dict) -> None:
    """Stopping a check stops the program it was waiting for, but not what THAT program started:
    stop `make test` and the test binary make launched runs on, at full speed, for ever. Stop every
    process that did not exist before the check (`before`, from _processes()) and either descends
    from this one or sits in its process group (an orphan: its parent is gone, so it now belongs to
    pid 1 or to a subreaper, whichever this machine has). Nothing that was already running is
    touched, the launcher that started this notebook included. Best effort."""
    now, me, group = _processes(), os.getpid(), os.getpgrp()
    if not before or not now:
        return
    def descends(pid):
        seen = set()
        while pid in now and pid not in seen:
            seen.add(pid)
            pid = now[pid][0]
            if pid == me:
                return True
        return False
    for pid, (_, pgid) in now.items():
        if pid != me and pid not in before and (pgid == group or descends(pid)):
            try:
                os.kill(pid, signal.SIGKILL)
            except OSError:
                pass
# --- END COMMONS TIME LIMIT ---


def _try(label: str, check: Callable[[], None], needs: tuple[str, ...] = ()) -> None:
    """Run a check, or a demo that depends on your code, without derailing the notebook.

    A stub you have not filled in yet simply says so. A wrong answer prints the check's own
    message -- which names the likely mistake -- and the notebook carries on. A demo names the
    exercises it `needs`: until each has passed its check, the demo says which one it is
    waiting for and skips.
    """
    waiting = [name for name in _EXERCISES
               if name in needs and _STATUS.get(name) != "passed"]
    if waiting:
        _STATUS[label] = "not started"
        print(f"{label}: skipped — needs {_named(waiting)} to pass first.")
        return
    try:
        _limited(label, check)
    except NotImplementedError as exc:
        _STATUS[label] = "not started"
        stub = traceback.extract_tb(exc.__traceback__)[-1].name
        owner = [name for name, funcs in _EXERCISES.items() if stub in funcs and name != label]
        if owner:
            print(f"{label}: skipped — needs {_named(owner)} first.")
        elif label in _EXERCISES:
            print(f"{label}: not implemented yet — fill in {stub}() above, then re-run "
                  "this cell.")
        else:
            print(f"{label}: skipped — {stub}() is not implemented yet.")
    except AssertionError as exc:
        _STATUS[label] = "failed"
        _FAILED_CHECKS.append(label)
        print(f"{label}: FAILED — {exc}")
    except Exception as exc:
        _STATUS[label] = "failed"
        _FAILED_CHECKS.append(label)
        print(f"{label}: raised {type(exc).__name__}: {exc}")
    else:
        _STATUS[label] = "passed"
```

## 3. Exercise 1 — `welch_psd()`

Detector noise is not flat: it is far louder at low frequencies, where ground motion (seismic
noise) dominates, and it has narrow spectral lines such as power-grid harmonics and
calibration lines (`claims.yaml`). Before anything else can be done,
you need an estimate of that noise's power spectral density (PSD) — Welch's method: split
the data into overlapping, windowed segments, periodogram each one, and average.

<details><summary>💡 Hint 1 — what to think about</summary>

A "periodogram" of one windowed segment is `|rfft(segment * window)|**2`, scaled by a
constant so that averaging many segments converges to a *density* (power per Hz), not a raw
spectrum. `np.fft.rfft` already gives you only the non-negative frequencies; `np.fft.rfftfreq`
gives you the matching frequency axis.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Scale each segment's `|rfft|**2` by `1 / (fs * sum(window**2))` — this is the standard
"density" normalisation (it is what makes the PSD's units power/Hz, and it is what section
5's whitening formula assumes). Step through the signal `nperseg - noverlap` samples at a
time, window and periodogram each segment, and average all of them elementwise. Then make it
one-sided: `rfft` keeps only the positive half of a spectrum whose power is split between
positive and negative frequencies, so double every bin except DC and (for an even `nperseg`)
Nyquist. Return the frequency axis and the PSD.
</details>

```
def hann_window(n: int) -> np.ndarray:
    """A length-`n` Hann window. Given to you — every exercise below that windows something
    uses this, so its normalisation is fixed once, here."""
    if n <= 1:
        return np.ones(n)
    k = np.arange(n)
    return 0.5 - 0.5 * np.cos(2 * np.pi * k / (n - 1))


def welch_psd(x: np.ndarray, fs: float, nperseg: int,
              noverlap: int | None = None) -> tuple[np.ndarray, np.ndarray]:
    """Welch's method: the averaged, windowed periodogram estimate of `x`'s one-sided PSD.

    `noverlap` defaults to `nperseg // 2` (50% overlap). Each segment is windowed with
    `hann_window(nperseg)` before its periodogram is taken; segments step through `x` by
    `nperseg - noverlap` samples, and any samples left over at the end (fewer than one more
    full segment) are simply not used. A periodogram is scaled by
    `1 / (fs * sum(window**2))` so that the average is a power *density* (per Hz), consistent
    across different window shapes and lengths. The result is ONE-SIDED: the power at negative
    frequencies is folded onto the positive ones, so every bin except DC (bin 0) and, for an
    even `nperseg`, the Nyquist bin (the last one) is doubled.

    Returns `(freqs, psd)`, both length `nperseg // 2 + 1`, matching `np.fft.rfftfreq(nperseg,
    d=1/fs)`.

    Example:
        >>> rng = np.random.default_rng(0)
        >>> x = rng.normal(0, 1.0, 8192)
        >>> freqs, psd = welch_psd(x, fs=1024.0, nperseg=512)
        >>> float(freqs[0]), float(freqs[-1])
        (0.0, 512.0)
        >>> bool(0.5 < psd.mean() * 1024 / 2 < 2.0)   # white noise: PSD ~ flat near 2*var/fs
        True
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_welch_psd() -> None:
    rng = np.random.default_rng(0)
    fs = 1024.0
    x = rng.normal(0, 1.0, 16384)
    freqs, psd = welch_psd(x, fs, nperseg=1024)
    assert freqs[0] == 0.0 and abs(freqs[-1] - fs / 2) < 1e-9, (
        f"got freqs from {freqs[0]} to {freqs[-1]} — expected 0 to fs/2 ({fs / 2}), matching "
        "np.fft.rfftfreq(nperseg, d=1/fs)"
    )
    assert len(freqs) == len(psd) == 1024 // 2 + 1, (
        f"got {len(psd)} PSD bins for nperseg=1024 — expected nperseg//2 + 1 = 513"
    )
    assert np.all(psd >= 0), "a power spectral density can never be negative"
    # unit-variance white noise: one-sided PSD integrates (0..fs/2) to the variance, so it
    # should sit close to 2 * var / fs = 2/1024 across the band (away from the DC/Nyquist bins).
    expected = 2.0 / fs
    mid = psd[10:-10]
    assert 0.5 * expected < mid.mean() < 2.0 * expected, (
        f"mean PSD (away from the edges) was {mid.mean():.2e}, expected close to "
        f"{expected:.2e} (= 2 * noise variance / fs) for unit-variance white noise — check "
        "the 1 / (fs * sum(window**2)) normalisation"
    )
    # more segments (finer step) should not change the ESTIMATE much, only its variance
    freqs2, psd2 = welch_psd(x, fs, nperseg=1024, noverlap=1024 - 64)
    assert abs(psd2[mid.argmax() + 10] - psd[mid.argmax() + 10]) < 5 * expected, (
        "changing noverlap changed the PSD estimate by far more than sampling noise explains "
        "— the periodogram normalisation should not depend on how much segments overlap"
    )
    # a pure sinusoid concentrates power at its own frequency
    t = np.arange(8192) / fs
    tone = 3.0 * np.sin(2 * np.pi * 200.0 * t)
    freqs3, psd3 = welch_psd(tone, fs, nperseg=1024)
    peak_freq = freqs3[np.argmax(psd3)]
    assert abs(peak_freq - 200.0) < freqs3[1] - freqs3[0] + 1e-9, (
        f"a 200 Hz tone's PSD peaked at {peak_freq} Hz — the peak bin should sit at the "
        "tone's own frequency, within one frequency-resolution step"
    )


_try("exercise 1", _check_welch_psd)
```

```
def _show_psd() -> None:
    nperseg = int(round(2.0 * FS))
    freqs_h1, psd_h1 = welch_psd(H1_STRAIN, FS, nperseg)
    freqs_l1, psd_l1 = welch_psd(L1_STRAIN, FS, nperseg)
    i10, i150 = np.searchsorted(freqs_h1, 10.0), np.searchsorted(freqs_h1, 150.0)
    print(f"frequency resolution: {freqs_h1[1] - freqs_h1[0]:.2f} Hz "
          f"({len(freqs_h1)} bins, 0 to {freqs_h1[-1]:.0f} Hz)")
    for name, psd in (("H1", psd_h1), ("L1", psd_l1)):
        print(f"{name} PSD @10 Hz: {psd[i10]:.3e} /Hz   @150 Hz: {psd[i150]:.3e} /Hz   "
              f"10 Hz is 10^{np.log10(psd[i10] / psd[i150]):.1f} times louder")
    globals().update(FREQS_H1=freqs_h1, PSD_H1=psd_h1, FREQS_L1=freqs_l1, PSD_L1=psd_l1)


_try("PSD demo", _show_psd, needs=("exercise 1",))
```

## 4. Exercise 2 — `whiten()`

The raw strain is dominated by low-frequency noise, many orders of magnitude louder (the demo
above measured by how much) than the part of the spectrum where the signal actually chirps.
**Whitening** divides the data, in the
frequency domain, by its own noise amplitude spectral density (the square root of the PSD),
so that — if the PSD estimate is a good description of the noise — every frequency bin ends
up contributing roughly equally. Section 9 measures, on this real data, what happens to a
matched filter that skips this step.

<details><summary>💡 Hint 1 — what to think about</summary>

`welch_psd` gives you a PSD on a coarse frequency grid (whatever `nperseg` gives). `whiten`
needs the PSD interpolated onto the FULL-length data's own `np.fft.rfftfreq(len(x), 1/fs)`
grid before dividing. `np.interp` does linear interpolation for you. The very first bin (DC,
frequency 0) is a special case Welch's PSD is undefined at usefully — reuse the next bin's
value there instead of interpolating into it.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

FFT the data. Interpolate `(freqs_psd, psd)` onto the data's own rfft frequency grid, then
fix up bin 0. Divide the data's FFT by `sqrt(psd_interp * fs / 2)` — this specific
normalisation is the one `welch_psd`'s own density scaling implies, and it is what makes
unit-variance white noise whiten to (approximately) unit-variance white noise. Inverse-FFT
back to a real time series of the same length as `x`.
</details>

```
def whiten(x: np.ndarray, freqs_psd: np.ndarray, psd: np.ndarray, fs: float) -> np.ndarray:
    """Divide `x`, in the frequency domain, by its own noise amplitude spectral density.

    `freqs_psd, psd` are a PSD estimate on a (possibly coarser) frequency grid, e.g. from
    `welch_psd`; this function interpolates it onto `x`'s own `np.fft.rfftfreq(len(x), 1/fs)`
    grid, then divides `rfft(x)` by `sqrt(psd_interp * fs / 2)` at every frequency and
    inverse-transforms back. Bin 0 (DC) is set equal to bin 1's value before dividing, since
    Welch's PSD estimate is not informative there.

    Example:
        >>> rng = np.random.default_rng(0)
        >>> fs = 1024.0
        >>> x = rng.normal(0, 3.0, 16384)          # noise with variance 9, not 1
        >>> freqs, psd = welch_psd(x, fs, nperseg=1024)
        >>> xw = whiten(x, freqs, psd, fs)
        >>> len(xw) == len(x)
        True
        >>> bool(0.5 < xw.std() < 2.0)              # whitened noise has ~unit variance
        True
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_whiten() -> None:
    rng = np.random.default_rng(1)
    fs = 2048.0
    n = 32768
    x = rng.normal(0, 5.0, n)   # variance 25, well away from 1
    freqs, psd = welch_psd(x, fs, nperseg=2048)
    xw = whiten(x, freqs, psd, fs)
    assert len(xw) == n, f"whiten changed the length from {n} to {len(xw)}"
    assert 0.4 < xw.std() < 2.5, (
        f"whitened white noise had std {xw.std():.3f} — whitening white noise (whose PSD "
        "estimate you fed straight back in) should land close to unit variance, whatever the "
        "INPUT variance was"
    )
    # colored noise: much louder below 50 Hz. After whitening, high and low bands should have
    # comparable power -- unlike the raw (unwhitened) data.
    t = np.arange(n) / fs
    loud_low = rng.normal(0, 1.0, n) + 20.0 * np.sin(2 * np.pi * 10.0 * t + rng.normal(0, 1, n))
    freqs2, psd2 = welch_psd(loud_low, fs, nperseg=2048)
    raw_fft = np.fft.rfft(loud_low)
    raw_freqs = np.fft.rfftfreq(n, d=1 / fs)
    low_power_raw = np.sum(np.abs(raw_fft[(raw_freqs > 5) & (raw_freqs < 15)]) ** 2)
    high_power_raw = np.sum(np.abs(raw_fft[(raw_freqs > 200) & (raw_freqs < 210)]) ** 2)
    loud_w = whiten(loud_low, freqs2, psd2, fs)
    w_fft = np.fft.rfft(loud_w)
    low_power_w = np.sum(np.abs(w_fft[(raw_freqs > 5) & (raw_freqs < 15)]) ** 2)
    high_power_w = np.sum(np.abs(w_fft[(raw_freqs > 200) & (raw_freqs < 210)]) ** 2)
    raw_ratio = low_power_raw / high_power_raw
    w_ratio = low_power_w / high_power_w
    assert w_ratio < raw_ratio / 5, (
        f"before whitening, the 10 Hz band carried {raw_ratio:.0f}x the power of the 205 Hz "
        f"band; after whitening it was still {w_ratio:.0f}x — whitening should divide out "
        "most of that imbalance, not leave it unchanged"
    )
    # dividing by a LARGER PSD must reduce the output, not increase it
    xw_small_psd = whiten(x, freqs, psd * 0.01, fs)
    assert xw_small_psd.std() > xw.std() * 3, (
        "dividing by a SMALLER psd should make the whitened series LOUDER, not quieter — "
        "check the formula divides by sqrt(psd), not multiplies"
    )
    # the DC (bin 0) estimate must be patched to bin 1's value before dividing, or an
    # unrepresentative DC PSD estimate blows up a signal with real power at DC.
    dc_test = whiten(np.full(4096, 5.0), np.array([0.0, 10.0, 500.0]), np.array([1e-8, 1.0, 1.0]),
                     1024.0)
    assert abs(dc_test.mean()) < 50.0, (
        f"whitening a constant (all-DC) signal against a psd whose bin 0 is 1e-8 but whose "
        f"bin 1 is 1.0 gave a whitened mean of {dc_test.mean():.1f} — patch psd_interp[0] to "
        "psd_interp[1] before dividing"
    )
    # the normalisation, pinned: white noise whitened against its EXACT one-sided PSD (flat at
    # 2 * variance / fs) must come back with unit variance, not 0.71 or 1.41.
    sigma = 3.0
    y = rng.normal(0, sigma, n)
    yw = whiten(y, np.linspace(0.0, fs / 2, 5), np.full(5, 2.0 * sigma ** 2 / fs), fs)
    assert abs(yw.std() - 1.0) < 0.05, (
        f"white noise whitened against its exact PSD (flat, 2 * variance / fs) had std "
        f"{yw.std():.3f}, expected 1.00 — divide rfft(x) by sqrt(psd * fs / 2); sqrt(psd * fs) "
        "gives 0.71 and sqrt(psd * fs / 4) gives 1.41"
    )


_try("exercise 2", _check_whiten)
```

## 5. Exercise 3 — `bandpass_fd()`

GW150914 swept upwards in frequency from 35 to 250 Hz, and the discovery paper displays it
band-passed to 35-350 Hz (`claims.yaml`), the band this lesson uses. Outside that band this
data holds noise the whitening filter cannot fully tame. A frequency-domain band-pass zeroes
out everything outside `[f_low, f_high]`, with a short raised-cosine taper at each edge so
the filter does not ring as badly as a hard cutoff would.

<details><summary>💡 Hint 1 — what to think about</summary>

Build a real-valued mask the same length as `np.fft.rfftfreq(len(x), 1/fs)`: 1 everywhere in
`[f_low, f_high]`, 0 well outside it, and a smooth `0.5 - 0.5*cos(...)` ramp of width `taper`
Hz just OUTSIDE each edge. Multiply the FFT by the mask and inverse-transform.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Start the mask at all-ones. Zero everything with frequency below `f_low` or above `f_high`.
Then overwrite the two `taper`-Hz-wide bands just outside each edge with a half-cosine ramp:
rising from 0 to 1 across `[f_low - taper, f_low]`, falling from 1 to 0 across
`[f_high, f_high + taper]`. Every frequency inside `[f_low, f_high]` keeps mask 1.
</details>

```
def bandpass_fd(x: np.ndarray, fs: float, f_low: float, f_high: float,
                 taper: float = 5.0) -> np.ndarray:
    """Zero out everything outside `[f_low, f_high]` Hz, with a `taper`-Hz raised-cosine edge.

    The gain is 1 at every frequency in `[f_low, f_high]` and 0 below `f_low - taper` or above
    `f_high + taper`. The two `taper`-Hz bands just OUTSIDE the pass band ramp smoothly: across
    `[f_low - taper, f_low]` the gain rises from 0 to 1 as
    `0.5 - 0.5*cos(pi * (f - (f_low - taper)) / taper)`, and across `[f_high, f_high + taper]`
    it falls from 1 to 0 as the mirror image. `taper=0` is a hard cutoff at `f_low` and `f_high`.

    Example:
        >>> fs = 1024.0
        >>> x = np.random.default_rng(0).normal(0, 1.0, 4096)
        >>> y = bandpass_fd(x, fs, 100.0, 200.0, taper=5.0)
        >>> len(y) == len(x)
        True
        >>> freqs = np.fft.rfftfreq(len(x), d=1/fs)
        >>> below = np.abs(np.fft.rfft(y))[freqs < 90.0]
        >>> float(below.max()) < 1e-6 * float(np.abs(np.fft.rfft(x)).max())
        True
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_bandpass_fd() -> None:
    fs = 1024.0
    n = 4096
    rng = np.random.default_rng(2)
    x = rng.normal(0, 1.0, n)
    y = bandpass_fd(x, fs, 100.0, 200.0, taper=5.0)
    assert len(y) == n, f"bandpass_fd changed the length from {n} to {len(y)}"
    freqs = np.fft.rfftfreq(n, d=1 / fs)
    Yf = np.fft.rfft(y)
    far_below = np.abs(Yf[freqs < 90.0])
    far_above = np.abs(Yf[freqs > 210.0])
    ref = np.abs(np.fft.rfft(x)).max()
    assert far_below.max() < 1e-6 * ref, (
        f"power below 90 Hz was not removed (max {far_below.max():.2e} vs reference "
        f"{ref:.2e}) — frequencies more than `taper` below f_low must be zeroed"
    )
    assert far_above.max() < 1e-6 * ref, "power above 210 Hz was not removed"
    passband = np.abs(Yf[(freqs > 140.0) & (freqs < 160.0)])
    passband_in = np.abs(np.fft.rfft(x)[(freqs > 140.0) & (freqs < 160.0)])
    assert np.allclose(passband, passband_in, rtol=1e-6), (
        "frequencies well inside [f_low, f_high] and away from either edge must pass through "
        "with gain 1 (unchanged), not be attenuated"
    )
    # a wider taper ramps further below f_low: 97 Hz sits inside a 5 Hz taper's ramp
    # ([95, 100]) but below a 1 Hz taper's ramp ([99, 100]), so it must pass THROUGH more
    # with the wider taper than with the narrow one.
    y_narrow_taper = bandpass_fd(x, fs, 100.0, 200.0, taper=1.0)
    y_wide_taper = bandpass_fd(x, fs, 100.0, 200.0, taper=5.0)
    i97 = np.searchsorted(freqs, 97.0)
    narrow_97 = np.abs(np.fft.rfft(y_narrow_taper)[i97])
    wide_97 = np.abs(np.fft.rfft(y_wide_taper)[i97])
    assert wide_97 > narrow_97 * 5, (
        f"at 97 Hz, a 1 Hz taper gave {narrow_97:.2e} and a 5 Hz taper gave {wide_97:.2e} — "
        "97 Hz is inside a 5 Hz taper's ramp band [95, 100] but below a 1 Hz taper's ramp "
        "band [99, 100], so the wider taper must pass noticeably more power through there"
    )
    # zero-width band: nothing at all should pass
    y_zero = bandpass_fd(np.ones(1024), fs, 500.0, 500.1, taper=0.01)
    assert np.abs(y_zero).max() < 1e-3, "an essentially empty pass band should pass essentially nothing"
    # ramp DIRECTION: the low edge must rise INTO the band, the high edge fall OUT of it —
    # not swapped.
    y_wide = bandpass_fd(x, fs, 100.0, 200.0, taper=10.0)
    Yw = np.fft.rfft(y_wide)
    low_start = abs(Yw[np.searchsorted(freqs, 91.0)])
    low_end = abs(Yw[np.searchsorted(freqs, 99.0)])
    assert low_end > low_start * 3, (
        "the LOW-edge ramp must rise TOWARDS f_low, not away from it — check the (freqs - "
        "(f_low - taper)) / taper argument and its sign"
    )
    high_start = abs(Yw[np.searchsorted(freqs, 201.0)])
    high_end = abs(Yw[np.searchsorted(freqs, 209.0)])
    assert high_start > high_end * 3, (
        "the HIGH-edge ramp must fall AWAY from f_high, not towards it"
    )
    # the exact ramp: an impulse has a flat spectrum, so the output's spectrum IS the mask.
    impulse = np.zeros(n)
    impulse[0] = 1.0
    gain = np.abs(np.fft.rfft(bandpass_fd(impulse, fs, 100.0, 200.0, taper=10.0)))
    for f0, want in ((85.0, 0.0), (95.0, 0.5), (100.0, 1.0), (200.0, 1.0), (205.0, 0.5),
                     (215.0, 0.0)):
        got = gain[np.searchsorted(freqs, f0)]
        assert abs(got - want) < 1e-6, (
            f"with f_low=100, f_high=200, taper=10 the gain at {f0:.0f} Hz was {got:.3f}, "
            f"expected {want} — the gain is 1 across the WHOLE of [f_low, f_high], and each "
            "ramp sits just outside its edge, half-way (0.5) at f_low - taper/2 and "
            "f_high + taper/2"
        )


_try("exercise 3", _check_bandpass_fd)
```

## 6. Exercise 4 — `chirp_frequency()`

A binary's gravitational-wave frequency rises as it spirals in. At leading (Newtonian,
quadrupole) order the instantaneous frequency `f(t)`, a time `t` before coalescence time
`tc`, depends on nothing but the **chirp mass** `M_c = (m1*m2)**(3/5) / (m1+m2)**(1/5)`
(Balasubramanian, Sathyaprakash & Dhurandhar 1995, eq. 2.4-2.5 — `claims.yaml`):

`f(t) = (1/pi) * (5 / (256 * (tc - t)))**(3/8) * (G*M_c/c^3)**(-5/8)`

valid for `t < tc` (it is undefined, not just large, at and after coalescence).

<details><summary>💡 Hint 1 — what to think about</summary>

Convert `mc_msun` (in solar masses) to `mc_msun * T_SUN` (a mass expressed as a time, in
seconds) before using it in the formula — every `G`, `c` in the equation above is already
folded into `T_SUN`. `tc - t` must be strictly positive; for `t >= tc` there is no frequency
to report (return `np.nan` there, elementwise, rather than raising or returning a complex
number from a negative base raised to a fractional power).
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

`tau = tc - t`. Build a boolean mask `tau > 0`. Where the mask is true, evaluate the formula
above with `mc_sec = mc_msun * T_SUN` in place of `G*M_c/c^3`; everywhere else, `nan`. The
frequency must be strictly INCREASING as `t` approaches `tc` — a real inspiral only ever
speeds up.
</details>

```
def chirp_frequency(mc_msun: float, tc: float, t: np.ndarray) -> np.ndarray:
    """Leading-order (Newtonian) gravitational-wave frequency, `t` seconds before coalescence.

    `mc_msun` is the chirp mass in solar masses; `tc` and `t` are in seconds (`t` may be an
    array). Returns `np.nan` wherever `t >= tc`.

    Example:
        >>> f = chirp_frequency(30.0, tc=1.0, t=np.array([0.5, 0.9, 0.99]))
        >>> bool(f[0] < f[1] < f[2])          # frequency rises as t -> tc
        True
        >>> bool(np.isnan(chirp_frequency(30.0, tc=1.0, t=np.array([1.0, 1.5]))).all())
        True
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_chirp_frequency() -> None:
    t = np.linspace(0.1, 0.99, 50)
    f = chirp_frequency(30.0, tc=1.0, t=t)
    assert np.all(np.diff(f) > 0), (
        "frequency must be STRICTLY INCREASING as t approaches tc — a real inspiral only "
        "speeds up. This is exactly what a wrong-signed (tc - t) vs (t - tc) mistake breaks."
    )
    assert np.all(np.isnan(chirp_frequency(30.0, tc=1.0, t=np.array([1.0, 1.2, 5.0])))), (
        "t >= tc has no defined pre-coalescence frequency and must be NaN, not a number"
    )
    # heavier chirp mass -> slower inspiral -> LOWER frequency at the same time-to-merger
    f_light = chirp_frequency(20.0, tc=1.0, t=np.array([0.9]))[0]
    f_heavy = chirp_frequency(60.0, tc=1.0, t=np.array([0.9]))[0]
    assert f_heavy < f_light, (
        f"at the same 0.1 s before merger, a 60 Msun chirp mass gave {f_heavy:.1f} Hz and a "
        f"20 Msun one gave {f_light:.1f} Hz — a HEAVIER system evolves more slowly and should "
        "be at a LOWER frequency this long before coalescence"
    )
    # known value: for M_c = 30 Msun, 0.05 s before coalescence, f is close to 55.4 Hz
    f_known = chirp_frequency(30.0, tc=1.0, t=np.array([0.95]))[0]
    assert abs(f_known - 55.4) / 55.4 < 0.02, (
        f"chirp_frequency(30.0, tc=1.0, t=[0.95]) gave {f_known:.1f} Hz, expected close to "
        "55.4 Hz — check the formula's exponents and the T_SUN conversion"
    )


_try("exercise 4", _check_chirp_frequency)
```

## 7. Exercise 5 — `chirp_phase()`

The matched filter needs the waveform's *phase*, not just its frequency: `f(t)` is the
time-derivative of the phase divided by `2*pi`. Integrating `chirp_frequency` (same
reference, eq. 2.3) gives a closed form:

`phi(t) = phi_c - (16*pi*K/5) * (tc - t)**(5/8)`, where `K = f(t) * (tc - t)**(3/8)`, which
does not depend on `t`.

<details><summary>💡 Hint 1 — what to think about</summary>

`K` above is exactly the same `(1/pi) * (5/256)**(3/8) * mc_sec**(-5/8)` constant that
`chirp_frequency` already computes as `f(t) * tau**(3/8)` — write it once, the same way, in
both functions. The single most common mistake here is a coefficient or sign error that
breaks the identity `d(phase)/dt / (2*pi) == chirp_frequency(t)` — the check below
differentiates your phase numerically and compares, so it will name that mistake.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Same `tau = tc - t` and the same positivity mask as `chirp_frequency`. Where `tau` is
positive, apply the closed form in the section text; everywhere else, `nan`. Do not call
`chirp_frequency` from inside this function — compute `K` directly, so a mistake in one does
not automatically hide inside the other.
</details>

```
def chirp_phase(mc_msun: float, tc: float, t: np.ndarray, phi_c: float = 0.0) -> np.ndarray:
    """Leading-order (Newtonian) gravitational-wave phase, `t` seconds before coalescence.

    Must satisfy `d(phase)/dt / (2*pi) == chirp_frequency(mc_msun, tc, t)` (checked below, and
    again independently by the hidden rubric).

    Example:
        >>> phi = chirp_phase(30.0, tc=1.0, t=np.array([0.5, 0.9]))
        >>> bool(phi[0] < phi[1])              # phase increases towards coalescence
        True
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_chirp_phase() -> None:
    t = np.linspace(0.1, 0.99, 50)
    phi = chirp_phase(30.0, tc=1.0, t=t)
    assert np.all(np.diff(phi) > 0), (
        "phase must be strictly increasing towards coalescence, exactly like the frequency it "
        "is the integral of"
    )
    assert np.all(np.isnan(chirp_phase(30.0, tc=1.0, t=np.array([1.0, 2.0])))), (
        "t >= tc must give NaN phase, matching chirp_frequency"
    )
    # phi_c shifts the whole curve by an additive constant, nothing else
    phi0 = chirp_phase(30.0, tc=1.0, t=t, phi_c=0.0)
    phi5 = chirp_phase(30.0, tc=1.0, t=t, phi_c=5.0)
    assert np.allclose(phi5 - phi0, 5.0), (
        "phi_c must shift the phase by exactly that additive constant everywhere"
    )
    # the deepest check: d(phase)/dt / (2*pi) must equal chirp_frequency's OWN output. A
    # coefficient slip (e.g. an extra factor of pi) can still pass every check above.
    t_fine = np.linspace(0.5, 0.9, 4000)
    dt_fine = t_fine[1] - t_fine[0]
    phi_fine = chirp_phase(30.0, tc=1.0, t=t_fine)
    f_from_phase = np.gradient(phi_fine, dt_fine) / (2 * np.pi)
    f_direct = chirp_frequency(30.0, tc=1.0, t=t_fine)
    rel_err = np.max(np.abs(f_from_phase[10:-10] - f_direct[10:-10]) / f_direct[10:-10])
    assert rel_err < 1e-3, (
        f"d(chirp_phase)/dt / (2*pi) disagreed with chirp_frequency by a relative "
        f"{rel_err:.2e} — the two must be consistent; check chirp_phase's leading "
        "coefficient (16*pi*K/5, not 16*pi**2*K/5 or similar)"
    )


_try("exercise 5", _check_chirp_phase)
```

Given: a quadrature template pair (`cos(phase)` and `sin(phase)`, each carrying the
leading-order amplitude `f(t)**(2/3)` — the same reference's eq. 2.1, `claims.yaml`), tapered
at both ends and restricted to `[f_low, f_high]` so it never claims validity outside the band
the detectors are actually sensitive in. Built from your `chirp_frequency`/`chirp_phase`, so
it only works once those do — the matched filter needs BOTH quadratures because the signal's
true phase offset at the detector is unknown; section 9 uses them together.

```
def build_quadrature_templates(mc_msun: float, tc: float, fs: float, f_low: float = 30.0,
                                 f_high: float = 400.0,
                                 duration: float = 2.0) -> tuple[np.ndarray, np.ndarray]:
    """(cos-phase, sin-phase) leading-order templates, amplitude ~ f(t)**(2/3), tapered."""
    n = int(round(duration * fs))
    t = tc - duration + np.arange(n) / fs
    f = chirp_frequency(mc_msun, tc, t)
    phi = chirp_phase(mc_msun, tc, t)
    valid = np.isfinite(f) & (f >= f_low) & (f <= f_high)
    amp = np.zeros(n)
    amp[valid] = f[valid] ** (2.0 / 3.0)
    hc = np.zeros(n)
    hs = np.zeros(n)
    hc[valid] = amp[valid] * np.cos(phi[valid])
    hs[valid] = amp[valid] * np.sin(phi[valid])
    idxs = np.where(valid)[0]
    if len(idxs) > 20:
        ramp_n = max(1, int(0.05 * len(idxs)))
        w = hann_window(2 * ramp_n)
        for arr in (hc, hs):
            arr[idxs[:ramp_n]] *= w[:ramp_n]
            arr[idxs[-ramp_n:]] *= w[ramp_n:]
    return hc, hs
```

## 8. Exercise 6 — `matched_filter_snr()`

The matched filter correlates whitened, band-passed data against a whitened, band-passed
template at every possible arrival time at once (an FFT does this in one pass), then
normalises by the template's own norm. Because the true signal's phase at arrival is
unknown, the standard fix is to correlate against BOTH quadratures and combine them in
quadrature — this is exactly why `build_quadrature_templates` hands you two templates, not
one.

<details><summary>💡 Hint 1 — what to think about</summary>

Multiplying two spectra and transforming back computes a (circular) convolution, and
convolving with a time-REVERSED copy of the template is the same as correlating with the
template itself. Reverse it, zero-pad it to the data's length, and index `k` of the result
is the inner product of `data` with the template ending exactly at sample `k`.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

For each of `hc`, `hs`: build a zero-padded, length-`len(data)` array whose first
`len(template)` samples are the template reversed (`template[::-1]`); correlate via the FFT
product above; divide by `sqrt(sum(hc**2))` (use the SAME normalising constant, from `hc`,
for both quadratures — they are close enough in norm for a many-cycle chirp that a separate
`hs` norm is not needed). Combine: `sqrt(zc**2 + zs**2)`.
</details>

```
def matched_filter_snr(data: np.ndarray, hc: np.ndarray, hs: np.ndarray) -> np.ndarray:
    """Phase-independent (quadrature) matched-filter SNR time series.

    `data` is whitened and band-passed; `hc, hs` are a `build_quadrature_templates` pair,
    already put through the SAME whitening/band-pass as `data`. Returns an array the same
    length as `data`; `result[k]` is the SNR of a template ending exactly at sample `k`.

    Example:
        >>> rng = np.random.default_rng(0)
        >>> data = rng.normal(0, 1.0, 4096)
        >>> hc = np.zeros(512); hc[100:110] = 1.0
        >>> hs = np.zeros(512); hs[100:110] = 0.0
        >>> snr = matched_filter_snr(data, hc, hs)
        >>> len(snr) == len(data)
        True
        >>> bool((snr >= 0).all())
        True
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_matched_filter_snr() -> None:
    n = 8192
    rng = np.random.default_rng(3)
    noise = rng.normal(0, 1.0, n)
    hc, hs = build_quadrature_templates(28.0, tc=1.0, fs=4096.0, duration=1.0)
    norm = np.sqrt(np.sum(hc ** 2))
    m = len(hc)
    start = 3000
    # inject at an UNKNOWN phase, so a correct solution needs both quadratures
    phase_offset = 1.1
    injected = hc * np.cos(phase_offset) - hs * np.sin(phase_offset)
    scale = 12.0 / norm
    data = noise.copy()
    data[start:start + m] += injected * scale
    snr = matched_filter_snr(data, hc, hs)
    assert len(snr) == n, f"matched_filter_snr changed the length ({len(snr)} vs {n})"
    assert np.all(snr >= -1e-9), "an SNR time series (a magnitude) can never be negative"
    peak_idx = int(np.argmax(snr))
    expected_idx = start + m - 1
    assert abs(peak_idx - expected_idx) <= 2, (
        f"peak at sample {peak_idx}, expected close to {expected_idx} (where the injected "
        "template ends) — check the template is time-REVERSED before FFT correlation"
    )
    assert 8.0 < snr[peak_idx] < 20.0, (
        f"peak SNR was {snr[peak_idx]:.2f}, expected roughly the injected 12.0 (some noise "
        "makes this inexact, but not by a factor of 2) — this uses an UNKNOWN phase offset, "
        "so a single-quadrature (cosine-only) filter will substantially undershoot here"
    )
    # a clean copy of the template itself (no noise) should score close to its own norm
    clean = np.zeros(n)
    clean[start:start + m] = hc
    snr_clean = matched_filter_snr(clean, hc, hs)
    assert abs(snr_clean[start + m - 1] - norm) / norm < 0.05, (
        f"a noiseless copy of hc scored {snr_clean[start + m - 1]:.2f}, expected close to its "
        f"own norm {norm:.2f} — the normalisation should make a perfect match's SNR equal to "
        "the template's own norm"
    )
    peak_clean = int(np.argmax(snr_clean))
    assert peak_clean == start + m - 1, (
        f"a noiseless copy of hc ending at sample {start + m - 1} peaked at sample {peak_clean} "
        "— result[k] must be the template ENDING at sample k; check where the reversed template "
        "sits in the zero-padded array (an off-by-one shifts every arrival time)"
    )


_try("exercise 6", _check_matched_filter_snr)
```

## 9. What whitening buys you — measured, not asserted

Run the same matched filter, at a fixed plausible chirp mass, on the whitened+band-passed H1
data AND on band-passed-but-**not whitened** H1 data, both searched over the same window.
Only whitening should put the loudest peak anywhere near the true event.

```
def _show_whitening_matters() -> None:
    hc, hs = build_quadrature_templates(30.0, tc=1.0, fs=FS, duration=2.0)
    hc = bandpass_fd(hc, FS, BANDPASS_LOW_HZ, BANDPASS_HIGH_HZ)
    hs = bandpass_fd(hs, FS, BANDPASS_LOW_HZ, BANDPASS_HIGH_HZ)
    crop = int(4.0 * FS)
    lo, hi = crop, N_SAMPLES - crop

    snr_whitened = matched_filter_snr(H1W, hc, hs)
    peak_w = lo + int(np.argmax(snr_whitened[lo:hi]))

    not_whitened = bandpass_fd(H1_STRAIN, FS, BANDPASS_LOW_HZ, BANDPASS_HIGH_HZ)
    snr_raw = matched_filter_snr(not_whitened, hc, hs)
    peak_raw = lo + int(np.argmax(snr_raw[lo:hi]))

    print(f"WITH whitening:    peak at t={peak_w / FS:.3f} s "
          f"(true event at t={T_EVENT:.1f} s, off by {abs(peak_w / FS - T_EVENT):.3f} s)")
    print(f"WITHOUT whitening: peak at t={peak_raw / FS:.3f} s "
          f"(off by {abs(peak_raw / FS - T_EVENT):.3f} s)")
    if abs(peak_w / FS - T_EVENT) < 1.0 and abs(peak_raw / FS - T_EVENT) > 1.0:
        print("-> whitening is what makes the peak land near the real event; without it, "
              "the search locks onto un-whitened low-frequency noise instead.")


_try("whitening-matters demo", _show_whitening_matters,
     needs=("exercise 1", "exercise 2", "exercise 3", "exercise 4", "exercise 5", "exercise 6"))
```

## 10. Exercise 7 — `estimate_chirp_mass()`

The true chirp mass is not known in advance. Search a grid of candidate chirp masses; for
each one, build its quadrature templates, band-pass them the same way as the data, run
`matched_filter_snr`, and keep the best (candidate, peak SNR, peak sample) seen so far — an
FFT correlation already searches every arrival time within one call, so no separate loop
over candidate merger times is needed.

<details><summary>💡 Hint 1 — what to think about</summary>

Only look at `snr[search_lo:search_hi]` when finding each candidate's peak — the first and
last few seconds of a whitened, band-passed array are dominated by filter-edge ringing (see
section 5's demo), and an unrestricted search finds THAT instead of the real signal.
`argmax` on the restricted slice gives an index relative to the slice; add `search_lo` back.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Track the best `(mc, snr, index)` seen. For each `mc` in `mc_grid`: build its quadrature
templates with `tc=1.0` (an arbitrary reference — only relative alignment within the
correlation matters), band-pass both to the DATA's band (`BANDPASS_LOW_HZ` to
`BANDPASS_HIGH_HZ`; `f_low`/`f_high` only say where the template itself is defined), run
`matched_filter_snr(data, hc, hs)`, restrict to `[search_lo, search_hi)`, and if its max
beats the best seen so far, keep it (remembering to add `search_lo` back onto the local
argmax).
</details>

```
def estimate_chirp_mass(data: np.ndarray, mc_grid: np.ndarray, fs: float, search_lo: int,
                         search_hi: int, f_low: float = 30.0, f_high: float = 400.0,
                         duration: float = 2.0) -> tuple[float, float, int]:
    """Grid-search chirp mass by maximising matched-filter SNR.

    For each `mc` in `mc_grid`: build its templates with
    `build_quadrature_templates(mc, 1.0, fs, f_low, f_high, duration)` (`tc=1.0` is an arbitrary
    reference: only alignment within the correlation matters), band-pass both with
    `bandpass_fd(..., BANDPASS_LOW_HZ, BANDPASS_HIGH_HZ)` so they match the data's own band, run
    `matched_filter_snr(data, hc, hs)`, and look for its peak only inside the half-open window
    `[search_lo, search_hi)`.

    Returns `(best_mc, best_snr, best_peak_index)`: the grid value whose in-window peak is
    highest, that peak's SNR, and its index into `data` (so it already includes `search_lo`).

    Example:
        >>> fs = 4096.0
        >>> hc, hs = build_quadrature_templates(30.0, 1.0, fs, duration=1.0)
        >>> hc = bandpass_fd(hc, fs, BANDPASS_LOW_HZ, BANDPASS_HIGH_HZ)
        >>> data = np.zeros(int(4 * fs)); data[8000:8000 + len(hc)] = 10.0 * hc
        >>> mc, snr, idx = estimate_chirp_mass(data, np.array([25.0, 30.0, 35.0]), fs,
        ...                                    search_lo=4096, search_hi=16000, duration=1.0)
        >>> mc, idx == 8000 + len(hc) - 1
        (30.0, True)
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_estimate_chirp_mass() -> None:
    fs = 4096.0
    n = int(6.0 * fs)
    rng = np.random.default_rng(4)
    noise = rng.normal(0, 1.0, n)
    true_mc = 32.0
    tc_true = 4.0
    hc_true, hs_true = build_quadrature_templates(true_mc, tc_true, fs, duration=1.0)
    hc_true = bandpass_fd(hc_true, fs, BANDPASS_LOW_HZ, BANDPASS_HIGH_HZ)
    hs_true = bandpass_fd(hs_true, fs, BANDPASS_LOW_HZ, BANDPASS_HIGH_HZ)
    norm = np.sqrt(np.sum(hc_true ** 2))
    m = len(hc_true)
    start = int(round((tc_true - 1.0) * fs))
    injected = hc_true * np.cos(0.7) - hs_true * np.sin(0.7)
    data = noise.copy()
    data[start:start + m] += injected * (14.0 / norm)

    mc_grid = np.arange(20.0, 45.001, 1.0)
    best_mc, best_snr, best_idx = estimate_chirp_mass(
        data, mc_grid, fs, search_lo=int(1.0 * fs), search_hi=int(5.5 * fs), duration=1.0)
    assert abs(best_mc - true_mc) <= 2.0, (
        f"recovered chirp mass {best_mc:.1f} Msun, injected {true_mc:.1f} Msun — should land "
        "on (or immediately next to) the injected grid point"
    )
    expected_idx = start + m - 1
    assert abs(best_idx - expected_idx) <= 40, (
        f"recovered peak index {best_idx}, expected within 40 samples of {expected_idx} — a "
        "grid search that lands one step away from the true chirp mass (a mismatched "
        "template) is expected to shift the peak by a handful of milliseconds, but not by "
        "much more than that"
    )
    assert best_snr > 8.0, f"recovered SNR {best_snr:.2f} — should clearly detect a 14-SNR injection"

    # noiseless: an on-grid, band-passed injection comes back EXACTLY: its own grid value, its
    # own end sample, and its injected SNR of 14.
    clean = np.zeros(n)
    clean[start:start + m] = injected * (14.0 / norm)
    mc0, snr0, idx0 = estimate_chirp_mass(
        clean, mc_grid, fs, search_lo=int(1.0 * fs), search_hi=int(5.5 * fs), duration=1.0)
    assert mc0 == true_mc and idx0 == start + m - 1, (
        f"a NOISELESS injection of chirp mass {true_mc:.1f} ending at sample {start + m - 1} came "
        f"back as {mc0} Msun at sample {idx0} — the index must be exact: argmax of the window "
        "slice plus search_lo, nothing more"
    )
    assert abs(snr0 - 14.0) < 0.2, (
        f"a noiseless SNR-14 injection came back with SNR {snr0:.2f} — band-pass each candidate "
        "template to BANDPASS_LOW_HZ-BANDPASS_HIGH_HZ, the data's own band; an un-band-passed "
        "template's norm counts power the data does not have"
    )

    # the search window must actually restrict the peak search
    _, _, idx_restricted = estimate_chirp_mass(
        data, mc_grid, fs, search_lo=0, search_hi=int(2.0 * fs), duration=1.0)
    assert idx_restricted < int(2.0 * fs), (
        "with search_hi=2.0*fs, the returned peak index must stay inside [search_lo, "
        "search_hi) — it must not report a peak outside the window it was told to search"
    )


_try("exercise 7", _check_estimate_chirp_mass)
```

## 11. Exercise 8 — `check_light_travel_consistency()`

Hanford and Livingston are far enough apart that a gravitational wave — which travels at
the speed of light — cannot arrive at one more than `LIGHT_TRAVEL_S` (10 ms; `claims.yaml`)
before the other. Any measured arrival-time difference bigger than that is not a real
coincident signal.

</details>

## 12. The detection

Everything above, run on the real committed data: search both detectors, measure the
arrival-time difference and check it against the light-travel bound, then compare against
the discovery paper's own published numbers.

```
def _run_detection() -> None:
    mc_grid = np.arange(15.0, 45.001, 0.5)
    crop = int(4.0 * FS)
    lo, hi = crop, N_SAMPLES - crop

    mc_h1, snr_h1, idx_h1 = estimate_chirp_mass(H1W, mc_grid, FS, lo, hi)
    mc_l1, snr_l1, idx_l1 = estimate_chirp_mass(L1W, mc_grid, FS, lo, hi)

    t_h1, t_l1 = idx_h1 / FS, idx_l1 / FS
    dt = t_h1 - t_l1
    consistent = check_light_travel_consistency(dt)
    combined_snr = float(np.hypot(snr_h1, snr_l1))

    print(f"H1: chirp mass {mc_h1:.1f} Msun · SNR {snr_h1:.2f} · "
          f"peak GPS {GPS_START + t_h1:.3f}")
    print(f"L1: chirp mass {mc_l1:.1f} Msun · SNR {snr_l1:.2f} · "
          f"peak GPS {GPS_START + t_l1:.3f}")
    print(f"combined SNR (H1, L1 in quadrature): {combined_snr:.2f}   "
          f"(discovery paper: {PAPER_SNR:.0f} -- claims.yaml)")
    print(f"H1 arrives {dt * 1000:+.2f} ms relative to L1   "
          f"(discovery paper: L1 first, H1 6.9 ms later -- claims.yaml)")
    print(f"within the {LIGHT_TRAVEL_S * 1000:.0f} ms light-travel bound: {consistent}")
    mean_mc = 0.5 * (mc_h1 + mc_l1)
    print(f"mean recovered chirp mass: {mean_mc:.1f} Msun   "
          f"(discovery paper: ~{PAPER_CHIRP_MASS_MSUN:.0f} Msun, detector frame -- claims.yaml)")
    gaps = []
    if combined_snr < PAPER_SNR:
        gaps.append(f"{PAPER_SNR - combined_snr:.1f} less combined SNR")
    if abs(mean_mc - PAPER_CHIRP_MASS_MSUN) > 2 * float(mc_grid[1] - mc_grid[0]):
        gaps.append(f"a chirp mass {mean_mc - PAPER_CHIRP_MASS_MSUN:+.1f} Msun away")
    if gaps:
        print()
        print("Measured gaps from the discovery paper: " + " and ".join(gaps) + ". This "
              "template is leading-order (Newtonian): no post-Newtonian phase corrections, no "
              "merger or ringdown, no spin. That mismatch is the standard explanation for gaps "
              "like these; this lesson measures the gaps, it does not test the explanation.")
    globals().update(MC_H1=mc_h1, MC_L1=mc_l1, SNR_H1=snr_h1, SNR_L1=snr_l1,
                     T_PEAK_H1=t_h1, T_PEAK_L1=t_l1, ARRIVAL_DT=dt,
                     LIGHT_TRAVEL_OK=consistent, COMBINED_SNR=combined_snr)


_try("detection", _run_detection, needs=tuple(_EXERCISES.keys()))
```
