# P01-L09 · The pack under inspection — the capstone

**You will build:** an inspection of a complete, already-assembled, already-attested
conformity pack, using the tools you built in modules 1 to 8; then the repair; then the
thing that actually decides whether the repair was honest — an audit that tells a fixed
check apart from a **silenced** one.

**Time:** ~120 minutes · **Runs on:** a laptop CPU, no download, no network
· **Prerequisites:** `T10-L01-ai-act-conformity-pack` and every module of this programme,
`P01-L01` through `P01-L08`. This is the capstone; it assumes all eight.

A provider hands you a pack for a fictional credit-scoring system. It is complete. Every
item the evidence table asks for is present, and the provider's own attestation says so:
twelve checks, twelve green. Six of the artefacts are wrong, in six ways the earlier
modules taught you to detect, and the attestation is green anyway — because the checks that
produced it were the provider's, and they were not these.

By the end you will be able to:

1. Run a suite over a pack and report a check nobody ran as **unverified**, not as passed.
2. Turn red results into findings ranked by how much of the evidence each puts in doubt.
3. Measure what a repair did to the evidence: what it added, changed, and **removed**.
4. Measure what it did to the checks: which were switched off, which tolerance widened.
5. Measure which numbers in the repaired pack no longer reproduce from what they cite.
6. Combine those three into a verdict per finding — **repaired**, **silenced** or
   **unfixed** — and run it against five repairs that go green and fix nothing.
7. Plan the honest repair as operations against the pack, and watch every check go green
   for a reason that survives the audit.

> **This is engineering, not legal advice.** The article numbers and the quoted wording are
> sourced in `claims.yaml` with their URLs and access dates. Every threshold, every
> operation name and the whole idea of grading a repair by whether it silenced a check are
> **this lesson's modelling choices**. They are not statements about what any authority
> would accept. For a real system, read the Official Journal text and take professional
> advice.

## 2. The toolkit, given back to you

Everything in this cell you wrote in an earlier module. It is repeated here, unchanged in
behaviour and unchanged in name, so this notebook runs on its own — a capstone that made
you re-derive `verify_chain` would be module 1 again. Read the names; you know them all.

The one deliberate change: `P01-L03` held its typed sources in an `Artefact` NamedTuple.
Here every artefact is a plain dict, because this lesson copies, hashes and diffs whole
packs, and a pack that is not JSON all the way down cannot be compared to itself.

```
def verify_chain(log: list, expected_head: str | None = None) -> dict:
    """P01-L01's chain verifier, returning a dict rather than a NamedTuple."""
    for i, entry in enumerate(log):
        if entry["seq"] != i:
            return {"intact": False, "first_bad_seq": i,
                    "reason": f"sequence number at index {i} is {entry['seq']}: an entry has "
                              "been deleted or reordered"}
        expected_prev = log[i - 1]["entry_hash"] if i else GENESIS_HASH
        if entry["prev_hash"] != expected_prev:
            return {"intact": False, "first_bad_seq": i,
                    "reason": f"prev_hash at index {i} does not match the entry before it"}
        recomputed = digest_of({"seq": entry["seq"], "prev_hash": entry["prev_hash"],
                                "event": entry["event"]})
        if recomputed != entry["entry_hash"]:
            return {"intact": False, "first_bad_seq": i,
                    "reason": f"entry_hash at index {i} does not match its own contents"}
    if expected_head is not None:
        head = log[-1]["entry_hash"] if log else GENESIS_HASH
        if head != expected_head:
            return {"intact": False, "first_bad_seq": None,
                    "reason": f"the chain is internally consistent but its head is "
                              f"{head[:16]}…, not the published {expected_head[:16]}… — the "
                              "log was rewritten"}
    return {"intact": True, "first_bad_seq": None, "reason": ""}


def missing_seq(log: list) -> list:
    """P01-L01's `retention_status()["missing_seq"]`: the holes between 0 and the largest."""
    seqs = [entry["seq"] for entry in log]
    return sorted(set(range(max(seqs) + 1)) - set(seqs)) if seqs else []


MISSING = object()   # P01-L03's sentinel: "no such path", as against "a path holding nothing"


def field_value(sources: dict, path: str) -> Any:
    """P01-L03's path resolver, over plain-dict artefacts."""
    if "." not in path:
        return MISSING
    name, _, rest = path.partition(".")
    artefact = sources.get(name)
    if not isinstance(artefact, dict):
        return MISSING
    node: Any = artefact.get("payload")
    for key in rest.split("."):
        if not isinstance(node, dict) or key not in node:
            return MISSING
        node = node[key]
    return node


def is_sourced(value: Any) -> bool:
    """P01-L03's emptiness test. 0 and False are measurements; "" and [] are absences."""
    if value is MISSING or value is None:
        return False
    if isinstance(value, str):
        return bool(value.strip())
    if isinstance(value, dict):
        return any(is_sourced(inner) for inner in value.values())
    if isinstance(value, (list, tuple, set, frozenset)):
        return any(is_sourced(inner) for inner in value)
    return True


def check_section(spec: dict, sources: dict) -> dict:
    """P01-L03's section checker: what is missing, what is present but empty."""
    missing, empty, artefacts = [], [], set()
    for path in spec["requires"]:
        value = field_value(sources, path)
        if value is MISSING:
            missing.append(path)
        elif not is_sourced(value):
            empty.append(path)
        else:
            artefacts.add(path.split(".")[0])
    return {"number": spec["number"], "title": spec["title"],
            "complete": not (missing or empty), "artefacts": sorted(artefacts),
            "missing": missing, "empty": empty}


class UndeclarableMetricError(ValueError):
    """P01-L06: a measurement that cannot honestly be written into the instructions."""


DECLARATION_KEYS = ("point", "lo", "hi", "level", "n", "n_boot", "method")


def declare(name: str, ci: dict, higher_is_better: bool = True) -> dict:
    """P01-L06's declaration: refuses a bare point estimate."""
    if not isinstance(ci, dict):
        raise UndeclarableMetricError(
            f"{name}: a declaration needs an interval, not {type(ci).__name__}")
    absent = [k for k in DECLARATION_KEYS if k not in ci]
    if absent:
        raise UndeclarableMetricError(
            f"{name}: cannot be declared without {', '.join(absent)} — a level, a sample size "
            "and a method are part of the claim, not decoration")
    point, lo, hi = ci["point"], ci["lo"], ci["hi"]
    for label, value in (("point", point), ("lo", lo), ("hi", hi)):
        if value is None or (isinstance(value, float) and math.isnan(value)):
            raise UndeclarableMetricError(f"{name}: {label} is {value!r}")
    point, lo, hi = float(point), float(lo), float(hi)
    if hi < lo:
        raise UndeclarableMetricError(f"{name}: hi {hi} is below lo {lo}")
    if not (lo - 1e-12 <= point <= hi + 1e-12):
        raise UndeclarableMetricError(
            f"{name}: point {point} sits outside its own interval [{lo}, {hi}]")
    return {"name": str(name), "point": point, "lo": lo, "hi": hi,
            "level": float(ci["level"]), "width": hi - lo,
            "declared": lo if higher_is_better else hi,
            "direction": "higher_is_better" if higher_is_better else "lower_is_better",
            "method": str(ci["method"]), "n": int(ci["n"]), "n_boot": int(ci["n_boot"])}


def accuracy(y_true: np.ndarray, y_pred: np.ndarray) -> np.ndarray:
    """P01-L06's accuracy, reducing over the last axis so a bootstrap matrix works."""
    return np.mean(np.asarray(y_true) == np.asarray(y_pred), axis=-1)


def recall(y_true: np.ndarray, y_pred: np.ndarray) -> np.ndarray:
    """P01-L06's recall, with 0/0 reported as nan rather than as zero."""
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    hits = np.sum((y_true == 1) & (y_pred == 1), axis=-1)
    positives = np.sum(y_true == 1, axis=-1)
    return np.divide(hits, positives, out=np.full(np.shape(hits), np.nan, dtype=float),
                     where=positives > 0)


def false_positive_rate(y_true: np.ndarray, y_pred: np.ndarray) -> np.ndarray:
    """P01-L06's false positive rate."""
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    false_alarms = np.sum((y_true == 0) & (y_pred == 1), axis=-1)
    negatives = np.sum(y_true == 0, axis=-1)
    return np.divide(false_alarms, negatives,
                     out=np.full(np.shape(false_alarms), np.nan, dtype=float),
                     where=negatives > 0)


METRICS = {"accuracy": accuracy, "recall": recall,
           "false_positive_rate": false_positive_rate}
HIGHER_IS_BETTER = {"accuracy": True, "recall": True, "false_positive_rate": False}


def bootstrap_ci(y_true, y_pred=None, metric: Callable = accuracy, n_boot: int = 400,
                 level: float = 0.95, seed: int = 90210) -> dict:
    """P01-L06's paired percentile bootstrap: one index vector, both arrays."""
    y_true = np.asarray(y_true)
    y_pred = y_true if y_pred is None else np.asarray(y_pred)
    n = int(y_true.shape[0])
    if n == 0 or int(n_boot) <= 0:
        return {"point": float("nan"), "lo": 0.0, "hi": 1.0, "level": float(level),
                "n": n, "n_boot": int(n_boot), "method": "percentile bootstrap"}
    rng = np.random.default_rng(seed)
    idx = rng.integers(0, n, size=(int(n_boot), n))
    replicates = np.asarray(metric(y_true[idx], y_pred[idx]), dtype=float)
    alpha = (1.0 - float(level)) / 2.0
    lo, hi = np.percentile(replicates, [100.0 * alpha, 100.0 * (1.0 - alpha)])
    return {"point": float(metric(y_true, y_pred)), "lo": float(lo), "hi": float(hi),
            "level": float(level), "n": n, "n_boot": int(n_boot),
            "method": "percentile bootstrap"}


def control_limits(baseline_errors: int, baseline_n: int, window_size: int,
                   k: float = 3.0) -> dict:
    """P01-L07's p-chart limits. Sigma is over the WINDOW, not over the baseline."""
    if baseline_n <= 0 or window_size <= 0:
        return {"centre": 0.0, "sigma": 0.0, "k": float(k), "ucl": 1.0, "lcl": 0.0,
                "window_size": int(window_size), "baseline_n": int(baseline_n)}
    centre = baseline_errors / baseline_n
    sigma = math.sqrt(centre * (1.0 - centre) / window_size)
    return {"centre": float(centre), "sigma": float(sigma), "k": float(k),
            "ucl": min(1.0, centre + k * sigma), "lcl": max(0.0, centre - k * sigma),
            "window_size": int(window_size), "baseline_n": int(baseline_n)}


def windowed_monitor(rows: list, limits: dict, window_size: int = 50,
                     min_run: int = 2) -> dict:
    """P01-L07's tumbling-window monitor. One alert per confirmed run, not per window."""
    ordered = sorted(rows, key=lambda row: row["seq"])
    n_windows = len(ordered) // window_size
    windows = []
    for index in range(n_windows):
        chunk = ordered[index * window_size:(index + 1) * window_size]
        errors = sum(1 for row in chunk if row["wrong"])
        rate = errors / window_size
        windows.append({"index": index, "n": window_size, "errors": errors, "rate": rate,
                        "first_seq": chunk[0]["seq"], "last_seq": chunk[-1]["seq"],
                        "outside": bool(rate > limits["ucl"] or rate < limits["lcl"])})
    alerts, run = [], 0
    for window in windows:
        if window["outside"]:
            run += 1
            if run == min_run:
                alerts.append(window["index"] - min_run + 1)
        else:
            run = 0
    return {"n_windows": n_windows, "dropped_rows": len(ordered) - n_windows * window_size,
            "windows": windows, "alerts": alerts,
            "verdict": "alert" if alerts else "in_control"}


def confirmed_windows(report: dict, min_run: int = 2) -> list:
    """The window indices inside a CONFIRMED excursion. P01-L07: one wobble is not an
    incident, so a lone out-of-limit window is not one either. Given to you; not graded."""
    out, run = [], []
    for window in report["windows"]:
        if window["outside"]:
            run.append(window["index"])
            continue
        if len(run) >= min_run:
            out.extend(run)
        run = []
    if len(run) >= min_run:
        out.extend(run)
    return out


REASON_BEFORE_PLACING = "before_placing_on_the_market"
REASON_INTENDED_PURPOSE = "intended_purpose_modified"
REASON_PRE_DETERMINED = "pre_determined_and_documented"
REASON_COMPLIANCE_AFFECTED = "chapter_iii_compliance_affected"
REASON_NO_EFFECT = "no_effect_on_compliance_or_purpose"


def is_substantial_modification(change: dict, plan: dict) -> dict:
    """P01-L08's Article 3(23) / 43(4) ladder, in its original order."""
    placed = parse_day(plan.get("placed_on_market"))
    on = parse_day(change.get("on"))
    if placed is not None and on is not None and on < placed:
        substantial, reason, basis = False, REASON_BEFORE_PLACING, "Article 43(4)"
    elif change.get("modifies_intended_purpose"):
        substantial, reason, basis = True, REASON_INTENDED_PURPOSE, "Article 3(23)"
    elif (change.get("change_id") in tuple(plan.get("pre_determined") or ())
          and plan.get("documented_in_annex_iv_2f")):
        substantial, reason, basis = False, REASON_PRE_DETERMINED, "Article 43(4)"
    elif change.get("affects_chapter_iii_compliance"):
        substantial, reason, basis = True, REASON_COMPLIANCE_AFFECTED, "Article 3(23)"
    else:
        substantial, reason, basis = False, REASON_NO_EFFECT, "Article 3(23)"
    return {"change_id": change.get("change_id"), "on": change.get("on"),
            "substantial": substantial, "reason": reason, "basis": basis}


print("toolkit ready: verify_chain, missing_seq, field_value, is_sourced, check_section, "
      "declare,\n  bootstrap_ci, control_limits, windowed_monitor, "
      "is_substantial_modification")
```

## 3. The pack

Nothing is loaded from disk and nothing is downloaded. The whole pack is constructed below
from one seed and a list of explicit counts, so two runs agree to the last digit and to the
last hash. It is **synthetic**: no real provider, no real reviewer, no real applicant, and
none of the numbers it prints is a claim about any real credit portfolio.

Six artefact families, in the shapes the earlier modules left them in:

* `log` — module 1's chained entries, with `cold_store` holding two nightly exports taken
  off the pack before a retention job ran;
* `sources` and `annex_iv` — module 3's typed artefacts and the document generated from
  them;
* `accuracy` — module 6's declarations, with the evaluation set they were measured on
  retained beside them;
* `oversight` — module 5's decisions and verifications;
* `monitoring` — module 7's feed and its control limits;
* `conformity` — module 8's change log, declaration, and the hash-chained `ledger` that
  says when each of those was sealed.

The provider's own attestation is in `attestation`. It is green.

```
PACK_SEED = 20260923
N_LOG_DECISIONS = 110          # 4 log entries each, plus one model_deployed
PURGE_FROM, PURGE_TO = 181, 194   # the seq range a retention job removed, inclusive
N_OVERSIGHT_DECISIONS = 1400
STAMP_REVIEWER = "u:pieter"
N_STAMPED = 400                # verifications this reviewer recorded inside one hour
STAMP_SECONDS = 8              # each of them, start to finish
STAMP_HOUR = "2026-08-11T09:00:00+00:00"
WINDOW_SIZE = 50
WINDOW_ERRORS = (3, 2, 4, 3, 3, 2, 4, 3, 3, 4, 2, 3, 3, 3, 4, 2,     # baseline, windows 0-15
                 3, 4, 2, 3, 3, 2, 4, 3, 3, 3,                        # windows 16-25
                 9, 9, 9, 9, 9,                                       # the degradation, 26-30
                 3, 3, 3)                                             # windows 31-33
BASELINE_WINDOWS = 16          # the plan declared windows 0..15 and sealed them
N_EVAL = 1200
N_BOOT = 400
BOOT_SEED = 90210

_RNG = np.random.default_rng(PACK_SEED)
_EPOCH = datetime(2026, 2, 1, 6, 0, 0, tzinfo=timezone.utc)


def _iso(moment: datetime) -> str:
    return moment.isoformat().replace("+00:00", "Z")


def _build_log() -> list:
    """Module 1's event stream for 110 decisions, chained the way P01-L01 chained it."""
    events = [{"ts": _iso(_EPOCH), "event_type": "model_deployed", "actor": "svc:deploy",
               "system_id": SYSTEM_ID,
               "payload": {"model_version": "v4.2.0", "deployed_by": "u:sanne"}}]
    for i in range(N_LOG_DECISIONS):
        did = f"D-{4000 + i}"
        start = _EPOCH + timedelta(days=1 + i, hours=int(_RNG.integers(0, 9)))
        score = float(np.round(_RNG.uniform(0.51, 0.99), 4))
        for offset, (kind, actor, payload) in enumerate((
                ("inference_requested", "svc:scoring",
                 {"model_version": "v4.2.0", "use_start": _iso(start),
                  "input_reference": f"app-{7000 + i}"}),
                ("match_found", "svc:scoring",
                 {"match_score": score, "reference_database": "bureau-nl-2026-03"}),
                ("human_verification", "ui:caseworker",
                 {"verifier_ids": [f"u:{'marlies' if i % 2 else 'tomasz'}"]}),
                ("decision_recorded", "svc:scoring",
                 {"outcome": "approve" if score > 0.7 else "decline",
                  "use_end": _iso(start + timedelta(minutes=4))}))):
            events.append({"ts": _iso(start + timedelta(minutes=offset)), "event_type": kind,
                           "actor": actor, "system_id": SYSTEM_ID, "decision_id": did,
                           "payload": payload})
    log, prev = [], GENESIS_HASH
    for seq, event in enumerate(events):
        entry = {"seq": seq, "event": event, "prev_hash": prev}
        entry["entry_hash"] = digest_of({"seq": seq, "prev_hash": prev, "event": event})
        prev = entry["entry_hash"]
        log.append(entry)
    return log


def _build_oversight() -> dict:
    """Module 5's decisions and verifications, one verification each."""
    decisions, verifications = [], []
    stamp_start = parse_ts(STAMP_HOUR)
    stamped = set(range(200, 200 + N_STAMPED))       # which decisions the stamp covers
    for i in range(N_OVERSIGHT_DECISIONS):
        did = f"DEC-{i:04d}"
        vid = f"V-{i:04d}"
        if i in stamped:
            k = i - 200
            began = stamp_start + timedelta(seconds=9 * k)
            reviewer, seconds = STAMP_REVIEWER, float(STAMP_SECONDS)
        else:
            began = _EPOCH + timedelta(days=60 + i // 24, minutes=13 * (i % 24))
            reviewer = ("u:marlies", "u:tomasz", "u:ines")[i % 3]
            seconds = float(np.round(np.exp(np.log(120.0) + _RNG.normal(0.0, 0.5)), 1))
        decisions.append({"decision_id": did, "standing": True, "outcome": "approve",
                          "verification_ids": [vid]})
        verifications.append({"verification_id": vid, "decision_id": did,
                              "reviewer": reviewer, "started_at": _iso(began),
                              "ended_at": _iso(began + timedelta(seconds=seconds)),
                              "seconds": seconds, "voided": False})
    return {"decisions": decisions, "verifications": verifications,
            "policy": {"basis": "Article 14(4)(d) — the reviewer must be able to disregard "
                                "the output, which takes longer than reading the id"}}


def _build_monitoring_rows() -> list:
    """Module 7's feed: 34 tumbling windows of 50 decisions, with a planted degradation."""
    rows, start = [], datetime(2026, 6, 10, 0, 0, 0, tzinfo=timezone.utc)
    for w, errors in enumerate(WINDOW_ERRORS):
        wrong_at = set(int(x) for x in _RNG.permutation(WINDOW_SIZE)[:errors])
        for r in range(WINDOW_SIZE):
            seq = w * WINDOW_SIZE + r
            rows.append({"seq": seq, "ts": _iso(start + timedelta(minutes=30 * seq)),
                         "wrong": r in wrong_at})
    return rows


def _build_evaluation() -> dict:
    """Module 6's retained evaluation set: the labels and predictions behind the metrics."""
    y_true = (_RNG.random(N_EVAL) < 0.42).astype(int)
    flip = _RNG.random(N_EVAL) < 0.10
    y_pred = np.where(flip, 1 - y_true, y_true)
    return {"y_true": [int(v) for v in y_true], "y_pred": [int(v) for v in y_pred],
            "collected_on": "2026-05-04",
            "note": "held-out evaluation split, retained so every declared interval can be "
                    "recomputed by whoever reads the pack"}


ANNEX_IV_SPECS = (
    {"number": "1", "title": "General description of the AI system",
     "requires": ("model_registry.intended_purpose", "model_registry.provider",
                  "model_registry.version")},
    {"number": "2", "title": "Elements of the system and the development process",
     "requires": ("model_registry.design_rationale", "dataset_manifest.datasets",
                  "dataset_manifest.labelling_procedure")},
    {"number": "3", "title": "Monitoring, functioning and control",
     "requires": ("control_design.capabilities_and_limitations",
                  "control_design.human_oversight_measures")},
    {"number": "4", "title": "Performance metrics",
     "requires": ("metrics.primary_metric", "metrics.primary_metric_value")},
    {"number": "5", "title": "Risk management system",
     "requires": ("risk_register.identified_risks", "risk_register.residual_risk_statement")},
    {"number": "6", "title": "Changes through the lifecycle",
     "requires": ("change_record.changes",)},
    {"number": "7", "title": "Harmonised standards applied",
     "requires": ("standards_record.applied",)},
    {"number": "8", "title": "The EU declaration of conformity",
     "requires": ("declaration_record.issued_on", "declaration_record.route")},
    {"number": "9", "title": "The post-market monitoring system",
     "requires": ("post_market_plan.method", "post_market_plan.baseline_window")},
)


def _build_sources() -> dict:
    """Module 3's typed artefacts, as plain dicts. `metrics` is the one with a hole in it."""
    def artefact(name, payload, days_ago):
        return {"name": name, "run_id": f"run-{name}-2026",
                "generated_at": _iso(_EPOCH + timedelta(days=200 - days_ago)),
                "payload": payload}
    return {
        "model_registry": artefact("model_registry", {
            "intended_purpose": "score consumer credit applications for a Dutch lender",
            "provider": "Northwind Analytics BV", "version": "v4.2.0",
            "design_rationale": "a four-feature linear scorer, chosen so every decision can "
                                "be explained to the applicant"}, 40),
        "dataset_manifest": artefact("dataset_manifest", {
            "datasets": ["applications-2024", "applications-2025"],
            "labelling_procedure": "repayment observed at 12 months"}, 38),
        "control_design": artefact("control_design", {
            "capabilities_and_limitations": "declines are advisory; a caseworker decides",
            "human_oversight_measures": "one reviewer per file, two for biometric onboarding"},
            36),
        # The hole. The metrics run produced a primary metric NAME and no value, and the
        # document generator emitted the section anyway.
        "metrics": artefact("metrics", {
            "primary_metric": "accuracy", "primary_metric_value": "",
            "primary_metric_value_source": ""}, 30),
        "risk_register": artefact("risk_register", {
            "identified_risks": ["thin-file applicants", "postcode proxy"],
            "residual_risk_statement": "accepted by the risk owner on 2026-05-18"}, 28),
        "change_record": artefact("change_record", {
            "changes": ["C-0027", "C-0031", "C-0033"]}, 20),
        "standards_record": artefact("standards_record", {
            "applied": ["EN ISO/IEC 42001:2023 (management system, partial)"]}, 18),
        "declaration_record": artefact("declaration_record", {
            "issued_on": "2026-08-14", "route": "annex_vi_internal_control"}, 12),
        "post_market_plan": artefact("post_market_plan", {
            "method": "p-chart over tumbling windows of 50 decisions",
            "baseline_window": "windows 0-15, sealed before the period they judge"}, 10),
    }


CHANGES = (
    {"change_id": "C-0021", "on": "2026-02-20", "description": "logging retention job",
     "affects_chapter_iii_compliance": False, "modifies_intended_purpose": False},
    {"change_id": "C-0024", "on": "2026-04-02", "description": "feature store upgrade",
     "affects_chapter_iii_compliance": False, "modifies_intended_purpose": False},
    {"change_id": "C-0027", "on": "2026-05-11", "description": "threshold retune, planned",
     "affects_chapter_iii_compliance": True, "modifies_intended_purpose": False},
    {"change_id": "C-0029", "on": "2026-06-18", "description": "UI copy",
     "affects_chapter_iii_compliance": False, "modifies_intended_purpose": False},
    # The one that matters: the system was pointed at a new population after the assessment
    # on file, and the declaration was issued afterwards as though nothing had happened.
    {"change_id": "C-0031", "on": "2026-07-02",
     "description": "extended from consumer credit to small-business lending",
     "affects_chapter_iii_compliance": True, "modifies_intended_purpose": True},
    {"change_id": "C-0033", "on": "2026-07-29", "description": "latency patch",
     "affects_chapter_iii_compliance": False, "modifies_intended_purpose": False},
    {"change_id": "C-0035", "on": "2026-08-30", "description": "dashboard colours",
     "affects_chapter_iii_compliance": False, "modifies_intended_purpose": False},
)

CHANGE_PLAN = {"placed_on_market": "2026-03-02",
               "pre_determined": ("C-0027",), "documented_in_annex_iv_2f": True}


def ledger_append(ledger: list, kind: str, payload: dict) -> dict:
    """Append one hash-chained entry to the pack's ledger. Given to you; not graded."""
    prev = ledger[-1]["digest"] if ledger else GENESIS_HASH
    entry = {"seq": len(ledger), "kind": str(kind), "payload": dict(payload), "prev": prev}
    entry["digest"] = digest_of({"seq": entry["seq"], "kind": entry["kind"],
                                 "payload": entry["payload"], "prev": prev})
    ledger.append(entry)
    return entry


def verify_ledger(ledger: list) -> dict:
    """Module 6's ledger integrity check, over the pack's one ledger. Given to you."""
    prev = GENESIS_HASH
    for i, entry in enumerate(ledger):
        if entry["seq"] != i or entry["prev"] != prev:
            return {"intact": False, "first_bad_seq": i, "reason": "sequence or prev broken"}
        recomputed = digest_of({"seq": entry["seq"], "kind": entry["kind"],
                                "payload": entry["payload"], "prev": entry["prev"]})
        if recomputed != entry["digest"]:
            return {"intact": False, "first_bad_seq": i,
                    "reason": "the payload was edited after it was sealed"}
        prev = entry["digest"]
    return {"intact": True, "first_bad_seq": None, "reason": ""}


def build_pack() -> dict:
    """Assemble the whole pack. Deterministic: same seed, same hashes, every run."""
    full_log = _build_log()
    purged = [copy.deepcopy(e) for e in full_log if PURGE_FROM <= e["seq"] <= PURGE_TO]
    kept = [e for e in full_log if not (PURGE_FROM <= e["seq"] <= PURGE_TO)]

    rows = _build_monitoring_rows()
    baseline_errors = sum(WINDOW_ERRORS[:BASELINE_WINDOWS])
    baseline_n = BASELINE_WINDOWS * WINDOW_SIZE
    # The limits actually in the pack: refitted over every window the feed has, which is to
    # say over a period that already contained the excursion they are supposed to catch.
    refit_errors, refit_n = sum(WINDOW_ERRORS), len(WINDOW_ERRORS) * WINDOW_SIZE
    refit = control_limits(refit_errors, refit_n, WINDOW_SIZE, 3.0)

    evaluation = _build_evaluation()
    y_true, y_pred = evaluation["y_true"], evaluation["y_pred"]
    declarations = {}
    for name in ("recall", "false_positive_rate"):
        ci = bootstrap_ci(y_true, y_pred, metric=METRICS[name], n_boot=N_BOOT,
                          seed=BOOT_SEED)
        declarations[name] = {**declare(name, ci, HIGHER_IS_BETTER[name]),
                              "seed": BOOT_SEED, "metric": name}
    # The defect: a point estimate, typed into the instructions for use with no interval,
    # no level, no sample size and no method.
    declarations["accuracy"] = {
        "name": "accuracy", "metric": "accuracy",
        "point": round(float(accuracy(np.array(y_true), np.array(y_pred))), 4),
        "note": "measured on the 2026-05-04 evaluation split"}

    ledger: list = []
    ledger_append(ledger, "assessment", {"route": "annex_vi_internal_control",
                                         "on": "2026-05-20", "basis": "Article 43(2)"})
    ledger_append(ledger, "baseline_sealed",
                  {"windows": [0, BASELINE_WINDOWS - 1], "errors": baseline_errors,
                   "n": baseline_n, "window_size": WINDOW_SIZE, "sealed_on": "2026-06-05"})
    ledger_append(ledger, "declaration_issued", {"on": "2026-08-14", "assessment_seq": 0})

    sources = _build_sources()
    annex_iv = {"generated_on": "2026-08-12", "sections": []}
    for spec in ANNEX_IV_SPECS:
        section = {"number": spec["number"], "title": spec["title"], "emitted": True,
                   "artefacts": sorted({p.split(".")[0] for p in spec["requires"]})}
        if spec["number"] == "4":
            section["value"] = field_value(sources, "metrics.primary_metric_value")
            section["source_path"] = "metrics.primary_metric_value"
            section["source_digest"] = digest_of(sources["metrics"]["payload"])
        annex_iv["sections"].append(section)

    pack = {
        "system_id": SYSTEM_ID,
        "log": kept,
        "log_anchor": full_log[-1]["entry_hash"],
        "cold_store": {
            # Two nightly exports. One of them holds the entries the retention job removed;
            # the other is a different night and holds entries the log still has.
            "export-2026-05-14": {"segment_id": "export-2026-05-14",
                                  "covers": [PURGE_FROM, PURGE_TO],
                                  "entries": purged},
            "export-2026-04-30": {"segment_id": "export-2026-04-30", "covers": [40, 51],
                                  "entries": [copy.deepcopy(e) for e in full_log[40:52]]},
        },
        "sources": sources,
        "annex_iv": annex_iv,
        "accuracy": {"evaluation": evaluation, "declarations": declarations},
        "oversight": _build_oversight(),
        "monitoring": {"rows": rows, "limits": refit, "incidents": [],
                       "fitted_over": [0, len(WINDOW_ERRORS) - 1]},
        "conformity": {"changes": [dict(c) for c in CHANGES], "plan": dict(CHANGE_PLAN),
                       "declaration": {"issued_on": "2026-08-14", "status": "in_force",
                                       "assessment_seq": 0,
                                       "route": "annex_vi_internal_control"}},
        "ledger": ledger,
        "repair_ledger": [],
        "attestation": {"attested_on": "2026-08-15", "by": "u:sanne",
                        "claim": "12 of 12 internal checks green",
                        "checks_run": 12, "checks_green": 12},
    }
    return pack


PACK = build_pack()
assert json.dumps(PACK)      # the whole pack is JSON: it can be copied, hashed and diffed

_n_log = len(PACK["log"])
_n_ver = len(PACK["oversight"]["verifications"])
print(f"pack for {PACK['system_id']} · {_n_log} log entries · "
      f"{len(PACK['annex_iv']['sections'])} Annex IV sections · "
      f"{len(PACK['accuracy']['declarations'])} declared metrics")
print(f"  {_n_ver} verifications · {len(PACK['monitoring']['rows'])} monitored decisions · "
      f"{len(PACK['conformity']['changes'])} logged changes · "
      f"{len(PACK['ledger'])} sealed ledger entries")
print(f"\nthe provider's attestation: {PACK['attestation']['claim']}, signed "
      f"{PACK['attestation']['attested_on']} by {PACK['attestation']['by']}")
```

## 4. The suite

Six checks, each one built out of the toolkit above. They are **given to you** — writing
them again is modules 1 to 8, and you have done that. What is not given to you is anything
that reads their output.

Two things about the suite's shape matter for everything that follows.

* Each check is a row in `SUITE` with an `enabled` flag and its own `params`. A tolerance
  that lives in a dict can be widened, and a widening that happens in a dict can be
  measured. A tolerance baked into a function body cannot.
* `PARAM_DIRECTION` records, for every parameter, which way makes the check **weaker**.
  Without it "the parameter changed" is all you can say, and tightening a threshold is not
  the same act as loosening one.

```
SUITE = {
    "log_integrity": {"enabled": True, "params": {}},
    "document_sourcing": {"enabled": True, "params": {}},
    "declared_accuracy": {"enabled": True, "params": {"min_interval_level": 0.95}},
    "oversight_plausibility": {"enabled": True,
                               "params": {"min_seconds": 10, "max_reviews_per_hour": 60}},
    "declaration_currency": {"enabled": True, "params": {}},
    "monitoring_limits": {"enabled": True, "params": {"k": 3.0}},
}

# Which direction of change makes each parameter WEAKER. "lower" means a smaller number lets
# more through; "higher" means a larger one does.
PARAM_DIRECTION = {"min_interval_level": "lower", "min_seconds": "lower",
                   "max_reviews_per_hour": "higher", "k": "higher"}


def _result(green: bool, reason_code: str, affected, population, **detail) -> dict:
    """One check's answer: a verdict, a reason code, and the numbers behind both."""
    return {"green": bool(green), "reason_code": reason_code,
            "affected": affected, "population": population, "detail": detail}


def check_log_integrity(pack: dict, params: dict) -> dict:
    """P01-L01: is the chain intact, does it match the published anchor, are there holes?"""
    log = pack["log"]
    holes = missing_seq(log)
    chain = verify_chain(log, pack.get("log_anchor"))
    population = len(log) + len(holes)
    if holes:
        return _result(False, "purged_window", len(holes), population, keys=[],
                       missing_seq=holes, first_bad_seq=chain["first_bad_seq"],
                       chain_reason=chain["reason"])
    if not chain["intact"]:
        code = "head_mismatch" if chain["first_bad_seq"] is None else "chain_broken"
        return _result(False, code, 1, population, keys=["anchor"],
                       chain_reason=chain["reason"],
                       first_bad_seq=chain["first_bad_seq"])
    return _result(True, "intact", 0, population, head=pack.get("log_anchor"))


def check_document_sourcing(pack: dict, params: dict) -> dict:
    """P01-L03: does every emitted section come from a source that says something?"""
    specs = {spec["number"]: spec for spec in ANNEX_IV_SPECS}
    sections = [s for s in pack["annex_iv"]["sections"] if s.get("emitted")]
    unsourced = []
    for section in sections:
        spec = specs.get(section["number"])
        if spec is None:
            continue
        status = check_section(spec, pack["sources"])
        if not status["complete"]:
            unsourced.append({"number": section["number"], "empty": status["empty"],
                              "missing": status["missing"]})
    if unsourced:
        keys = [f"section:{u['number']}" for u in unsourced]
        keys += [f"artefact:{path.split('.')[0]}" for u in unsourced
                 for path in u["empty"] + u["missing"]]
        return _result(False, "unsourced_section", len(unsourced), len(sections),
                       keys=sorted(set(keys)),
                       sections=[u["number"] for u in unsourced], detail_rows=unsourced)
    return _result(True, "sourced", 0, len(sections))


def check_declared_accuracy(pack: dict, params: dict) -> dict:
    """P01-L06: does every declared metric carry an interval, at the declared level?"""
    declarations = pack["accuracy"]["declarations"]
    undeclarable, thin = [], []
    for name in sorted(declarations):
        try:
            declared = declare(name, declarations[name],
                               HIGHER_IS_BETTER.get(name, True))
        except UndeclarableMetricError as exc:
            undeclarable.append({"metric": name, "why": str(exc)})
            continue
        if declared["level"] < params["min_interval_level"] - 1e-12:
            thin.append({"metric": name, "level": declared["level"]})
    population = len(declarations)
    if undeclarable:
        return _result(False, "no_interval", len(undeclarable), population,
                       keys=[f"declaration:{u['metric']}" for u in undeclarable],
                       metrics=[u["metric"] for u in undeclarable], rows=undeclarable)
    if thin:
        return _result(False, "interval_below_declared_level", len(thin), population,
                       keys=[f"declaration:{t['metric']}" for t in thin],
                       metrics=[t["metric"] for t in thin], rows=thin,
                       min_interval_level=params["min_interval_level"])
    return _result(True, "declared", 0, population, metrics=sorted(declarations))


def check_oversight_plausibility(pack: dict, params: dict) -> dict:
    """P01-L05, pointed at throughput: could the recorded reviews have happened?"""
    decisions = {d["decision_id"]: d for d in pack["oversight"]["decisions"]}
    standing_v = [v for v in pack["oversight"]["verifications"]
                  if not v.get("voided")
                  and decisions.get(v["decision_id"], {}).get("standing")]
    too_fast = [v["verification_id"] for v in standing_v
                if float(v["seconds"]) < params["min_seconds"]]
    per_hour: dict = {}
    for v in standing_v:
        key = (v["reviewer"], v["started_at"][:13])
        per_hour.setdefault(key, []).append(v["verification_id"])
    over = {f"{r} {h}": ids for (r, h), ids in per_hour.items()
            if len(ids) > params["max_reviews_per_hour"]}
    implausible = sorted(set(too_fast) | {i for ids in over.values() for i in ids})
    population = len(decisions)
    if implausible:
        by_id = {v["verification_id"]: v for v in standing_v}
        keys = ([f"verification:{i}" for i in implausible]
                + [f"decision:{by_id[i]['decision_id']}" for i in implausible])
        return _result(False, "implausible_review", len(implausible), population,
                       keys=sorted(set(keys)),
                       n_implausible=len(implausible), too_fast=len(too_fast),
                       busiest_hour=max((len(v) for v in per_hour.values()), default=0),
                       hours_over_limit=sorted(over))
    unverified = sorted(did for did, d in decisions.items()
                        if d.get("standing")
                        and not [v for v in standing_v if v["decision_id"] == did])
    if unverified:
        return _result(False, "standing_action_without_verification", len(unverified),
                       population, keys=[f"decision:{d}" for d in unverified],
                       n_unverified=len(unverified))
    return _result(True, "plausible", 0, population,
                   busiest_hour=max((len(v) for v in per_hour.values()), default=0))


def check_declaration_currency(pack: dict, params: dict) -> dict:
    """P01-L08: does the declaration in force post-date the last substantial modification?"""
    conformity = pack["conformity"]
    changes = conformity["changes"]
    verdicts = [is_substantial_modification(c, conformity["plan"]) for c in changes]
    substantial = [v for v in verdicts if v["substantial"]]
    declaration = conformity.get("declaration") or {}
    population = len(changes)
    if declaration.get("status") != "in_force":
        return _result(False, "no_declaration_in_force", population, population,
                       keys=["declaration"], status=declaration.get("status"))
    ledger = pack["ledger"]
    seq = declaration.get("assessment_seq")
    entry = ledger[seq] if isinstance(seq, int) and 0 <= seq < len(ledger) else None
    if entry is None or entry["kind"] != "assessment":
        return _result(False, "declaration_without_assessment", population, population,
                       keys=["declaration"], assessment_seq=seq)
    assessed_on = parse_day(entry["payload"].get("on"))
    late = [v for v in substantial if (parse_day(v["on"]) or date.min) > assessed_on]
    if late:
        return _result(False, "declaration_predates_substantial_modification", len(late),
                       population,
                       keys=sorted(["declaration"]
                                   + [f"change:{v['change_id']}" for v in late]),
                       change_ids=[v["change_id"] for v in late],
                       latest_substantial=max(v["on"] for v in late),
                       assessed_on=entry["payload"].get("on"),
                       issued_on=declaration.get("issued_on"))
    if (parse_day(declaration.get("issued_on")) or date.min) < assessed_on:
        return _result(False, "declaration_predates_its_own_assessment", 1, population,
                       keys=["declaration"], issued_on=declaration.get("issued_on"),
                       assessed_on=entry["payload"].get("on"))
    return _result(True, "current", 0, population,
                   substantial=[v["change_id"] for v in substantial],
                   assessed_on=entry["payload"].get("on"))


def sealed_baseline(pack: dict) -> dict | None:
    """The first `baseline_sealed` entry in the ledger. Given to you; not graded."""
    for entry in pack["ledger"]:
        if entry["kind"] == "baseline_sealed":
            return entry
    return None


def check_monitoring_limits(pack: dict, params: dict) -> dict:
    """P01-L07: were the limits fitted to the sealed baseline, and is every alert filed?"""
    monitoring = pack["monitoring"]
    rows, limits = monitoring["rows"], monitoring["limits"]
    entry = sealed_baseline(pack)
    n_windows = len(rows) // WINDOW_SIZE
    if entry is None:
        return _result(False, "no_sealed_baseline", n_windows, n_windows, keys=["limits"])
    payload = entry["payload"]
    expected = control_limits(payload["errors"], payload["n"], payload["window_size"],
                              params["k"])
    from_sealed = windowed_monitor(rows, expected, WINDOW_SIZE)
    as_fitted = windowed_monitor(rows, limits, WINDOW_SIZE)
    differ = [w["index"] for w, s in zip(as_fitted["windows"], from_sealed["windows"])
              if w["outside"] != s["outside"]]
    if differ:
        return _result(False, "limits_not_from_sealed_baseline", len(differ), n_windows,
                       keys=["limits"], windows_hidden=differ, ucl_in_pack=limits["ucl"],
                       ucl_from_sealed=expected["ucl"],
                       alerts_from_sealed=from_sealed["alerts"])
    covered = {int(i) for inc in monitoring.get("incidents", [])
               for i in inc.get("windows", [])}
    uncovered = [i for i in confirmed_windows(as_fitted) if i not in covered]
    if uncovered:
        return _result(False, "unacknowledged_excursion", len(uncovered), n_windows,
                       keys=["limits"], windows_uncovered=uncovered,
                       alerts=as_fitted["alerts"])
    return _result(True, "in_control", 0, n_windows, alerts=as_fitted["alerts"],
                   incidents=len(monitoring.get("incidents", [])))


CHECKS = {"log_integrity": check_log_integrity,
          "document_sourcing": check_document_sourcing,
          "declared_accuracy": check_declared_accuracy,
          "oversight_plausibility": check_oversight_plausibility,
          "declaration_currency": check_declaration_currency,
          "monitoring_limits": check_monitoring_limits}

for _cid, _fn in CHECKS.items():
    _res = _fn(PACK, SUITE[_cid]["params"])
    print(f"  {_cid:24s} {'green' if _res['green'] else 'RED  '}  "
          f"{_res['reason_code']:42s} {_res['affected']}/{_res['population']}")
```

## 5. Two more things the pack comes with

**An evidence index.** Every artefact family, reduced to `{key: digest}`. The key names one
piece of evidence and the digest is its content, so comparing two indices tells you what a
repair added, what it changed, and — the interesting one — what it removed. A log entry is
keyed on the digest of its *event*, not on its sequence number, because a repair that
re-chains the log renumbers everything: a seq-keyed index would report the renumbering as
deletions and additions that never happened, while an event-keyed one reports what the
repair really did — nothing removed, and every entry after the hole rewritten. The table in
section 12 counts that for you.

**An operations vocabulary.** A repair here is not free-hand editing; it is a list of
operations, applied in order to a copy of the pack by `apply_repairs`. Ten of the
seventeen restore, recompute or re-seal something and record where the new value came
from. Seven of them change nothing about the world and everything about the check —
deleting the records, widening a tolerance, switching a check off, re-chaining a log,
republishing an anchor, editing a sealed ledger entry, or writing a value in by hand with
no source at all. They are in the vocabulary on purpose. You are going to need to catch
them.

```
def evidence_index(pack: dict) -> dict:
    """Reduce a pack to {family: {key: digest}}. Given to you; not graded."""
    index: dict = {}
    index["log"] = {f"event:{digest_of(e['event'])[:16]}": digest_of(e)
                    for e in pack["log"]}
    # The published head belongs to the evidence: a log re-chained around a hole is green
    # against a re-published anchor and against nothing else.
    index["log"]["anchor"] = str(pack.get("log_anchor"))
    index["annex_iv"] = {f"section:{s['number']}": digest_of(s)
                         for s in pack["annex_iv"]["sections"]}
    index["sources"] = {f"artefact:{name}": digest_of(a["payload"])
                        for name, a in pack["sources"].items()}
    accuracy_part = pack["accuracy"]
    index["accuracy"] = {f"declaration:{n}": digest_of(d)
                         for n, d in accuracy_part["declarations"].items()}
    index["accuracy"]["evaluation"] = digest_of([accuracy_part["evaluation"]["y_true"],
                                                 accuracy_part["evaluation"]["y_pred"]])
    oversight = pack["oversight"]
    index["oversight"] = {f"verification:{v['verification_id']}": digest_of(v)
                          for v in oversight["verifications"]}
    index["oversight"].update({f"decision:{d['decision_id']}": digest_of(d)
                               for d in oversight["decisions"]})
    monitoring = pack["monitoring"]
    index["monitoring"] = {f"row:{r['seq']}": digest_of(r) for r in monitoring["rows"]}
    index["monitoring"]["limits"] = digest_of(monitoring["limits"])
    index["monitoring"].update({f"incident:{i['incident_id']}": digest_of(i)
                                for i in monitoring.get("incidents", [])})
    conformity = pack["conformity"]
    index["conformity"] = {f"change:{c['change_id']}": digest_of(c)
                           for c in conformity["changes"]}
    index["conformity"]["declaration"] = digest_of(conformity.get("declaration"))
    index["conformity"].update({f"ledger:{e['seq']}": digest_of(e) for e in pack["ledger"]})
    return index


def _walk(pack: dict, path: str):
    """Resolve a dotted path to (container, last_key). Given to you; not graded."""
    parts = path.split(".")
    node = pack
    for key in parts[:-1]:
        node = node[int(key)] if isinstance(node, list) else node[key]
    return node, parts[-1]


def _op_restore_log_segment(pack, suite, op):
    segment = pack["cold_store"][op["segment"]]
    present = {e["seq"] for e in pack["log"]}
    restored = [copy.deepcopy(e) for e in segment["entries"] if e["seq"] not in present]
    pack["log"] = sorted(pack["log"] + restored, key=lambda e: e["seq"])
    return {"keys": [f"event:{digest_of(e['event'])[:16]}" for e in restored],
            "source": f"cold_store:{segment['segment_id']}",
            "source_digest": digest_of(segment["entries"]), "authority": None}


def _op_recompute_declaration(pack, suite, op):
    metric = op["metric"]
    evaluation = pack["accuracy"]["evaluation"]
    n_boot, seed = int(op.get("n_boot", N_BOOT)), int(op.get("seed", BOOT_SEED))
    ci = bootstrap_ci(evaluation["y_true"], evaluation["y_pred"],
                      metric=METRICS[metric], n_boot=n_boot, seed=seed)
    declaration = {**declare(metric, ci, HIGHER_IS_BETTER.get(metric, True)),
                   "seed": seed, "metric": metric}
    pack["accuracy"]["declarations"][metric] = declaration
    ledger_append(pack["ledger"], "measurement",
                  {"metric": metric, "digest": digest_of(declaration),
                   "on": AS_OF.isoformat()})
    return {"keys": [f"declaration:{metric}"], "source": "accuracy.evaluation",
            "source_digest": digest_of([evaluation["y_true"], evaluation["y_pred"]]),
            "authority": None}


def _op_resource_section(pack, suite, op):
    declaration = pack["accuracy"]["declarations"][op["from_declaration"]]
    artefact = pack["sources"][op["artefact"]]
    field = op["field"]
    artefact["payload"][field] = round(float(declaration["declared"]), 4)
    artefact["payload"][f"{field}_source"] = digest_of(declaration)
    keys = [f"artefact:{op['artefact']}"]
    for section in pack["annex_iv"]["sections"]:
        if section["number"] == op["section"]:
            section["value"] = artefact["payload"][field]
            section["source_digest"] = digest_of(artefact["payload"])
            keys.append(f"section:{section['number']}")
    return {"keys": keys, "source": f"accuracy.declarations.{op['from_declaration']}",
            "source_digest": digest_of(declaration), "authority": None}


def _op_void_verifications(pack, suite, op):
    authority, reason = str(op.get("authority", "")), str(op.get("reason", ""))
    if not authority.strip() or not reason.strip():
        raise ValueError("void_verifications needs an authority and a reason: a review "
                         "voided by nobody, for nothing, is a deletion with extra steps")
    wanted = set(op["ids"])
    touched = []
    for verification in pack["oversight"]["verifications"]:
        if verification["verification_id"] in wanted:
            verification.update({"voided": True, "voided_by": authority,
                                 "voided_reason": reason, "voided_on": AS_OF.isoformat()})
            touched.append(f"verification:{verification['verification_id']}")
    return {"keys": touched, "source": f"finding:{op.get('finding')}",
            "source_digest": None, "authority": authority}


def _op_withhold_decisions(pack, suite, op):
    reason = str(op.get("reason", ""))
    if not reason.strip():
        raise ValueError("withhold_decisions needs a reason")
    wanted = set(op["ids"])
    touched = []
    for decision in pack["oversight"]["decisions"]:
        if decision["decision_id"] in wanted:
            decision.update({"standing": False, "withheld_reason": reason,
                             "withheld_on": AS_OF.isoformat()})
            touched.append(f"decision:{decision['decision_id']}")
    return {"keys": touched, "source": f"finding:{op.get('finding')}",
            "source_digest": None, "authority": None}


def _latest_substantial(pack) -> date | None:
    days = [parse_day(v["on"])
            for v in (is_substantial_modification(c, pack["conformity"]["plan"])
                      for c in pack["conformity"]["changes"]) if v["substantial"]]
    return max([d for d in days if d], default=None)


def _op_reassess_conformity(pack, suite, op):
    on = parse_day(op.get("on"))
    latest = _latest_substantial(pack)
    if on is None:
        raise ValueError("reassess_conformity needs a date")
    if latest is not None and on < latest:
        raise ValueError(
            f"a reassessment dated {on} cannot cover a substantial modification made on "
            f"{latest}: that is the assessment the pack already had")
    entry = ledger_append(pack["ledger"], "assessment",
                          {"route": pack["conformity"]["declaration"].get(
                              "route", "annex_vi_internal_control"),
                           "on": on.isoformat(),
                           "basis": str(op.get("basis", "Article 43(4)"))})
    return {"keys": [f"ledger:{entry['seq']}"], "source": "conformity.changes",
            "source_digest": digest_of(pack["conformity"]["changes"]), "authority": None}


def _op_withdraw_declaration(pack, suite, op):
    reason = str(op.get("reason", ""))
    if not reason.strip():
        raise ValueError("withdraw_declaration needs a reason")
    pack["conformity"]["declaration"]["status"] = "withdrawn"
    entry = ledger_append(pack["ledger"], "declaration_withdrawn",
                          {"reason": reason, "on": AS_OF.isoformat()})
    return {"keys": ["declaration", f"ledger:{entry['seq']}"],
            "source": f"finding:{op.get('finding')}", "source_digest": None,
            "authority": None}


def _op_reissue_declaration(pack, suite, op):
    declaration = pack["conformity"]["declaration"]
    if declaration.get("status") != "withdrawn":
        raise ValueError("reissue_declaration before the old declaration was withdrawn: "
                         "two declarations in force is not a repair")
    assessments = [e for e in pack["ledger"] if e["kind"] == "assessment"]
    latest = assessments[-1]
    on = parse_day(op.get("on"))
    assessed_on = parse_day(latest["payload"]["on"])
    if on is None or assessed_on is None or on < assessed_on:
        raise ValueError(f"a declaration dated {op.get('on')} cannot rest on an assessment "
                         f"dated {latest['payload']['on']}")
    pack["conformity"]["declaration"] = {"issued_on": on.isoformat(), "status": "in_force",
                                         "assessment_seq": latest["seq"],
                                         "route": declaration.get("route")}
    entry = ledger_append(pack["ledger"], "declaration_issued",
                          {"on": on.isoformat(), "assessment_seq": latest["seq"]})
    return {"keys": ["declaration", f"ledger:{entry['seq']}"],
            "source": f"ledger:{latest['seq']}", "source_digest": latest["digest"],
            "authority": None}


def _op_refit_limits_to_sealed_baseline(pack, suite, op):
    entry = sealed_baseline(pack)
    payload = entry["payload"]
    k = suite["monitoring_limits"]["params"]["k"]
    pack["monitoring"]["limits"] = control_limits(payload["errors"], payload["n"],
                                                  payload["window_size"], k)
    pack["monitoring"]["fitted_over"] = list(payload["windows"])
    return {"keys": ["limits"], "source": f"ledger:{entry['seq']}",
            "source_digest": entry["digest"], "authority": None}


def _op_file_incident(pack, suite, op):
    windows = sorted(int(w) for w in op.get("windows", []))
    if not windows:
        raise ValueError("file_incident needs the window indices the incident covers")
    incident = {"incident_id": str(op["incident_id"]), "aware_on": str(op["aware_on"]),
                "windows": windows, "article": str(op.get("article", "Article 73(1)")),
                "summary": str(op.get("summary", "excursion found on inspection"))}
    pack["monitoring"]["incidents"].append(incident)
    entry = ledger_append(pack["ledger"], "incident_filed", dict(incident))
    return {"keys": [f"incident:{incident['incident_id']}", f"ledger:{entry['seq']}"],
            "source": "monitoring.rows", "source_digest": None, "authority": None}


_DELETE_TARGETS = {
    "log": ("log", "seq"),
    "oversight_verifications": ("oversight.verifications", "verification_id"),
    "oversight_decisions": ("oversight.decisions", "decision_id"),
    "conformity_changes": ("conformity.changes", "change_id"),
    "monitoring_rows": ("monitoring.rows", "seq"),
    "annex_iv_sections": ("annex_iv.sections", "number"),
}


def _op_delete_records(pack, suite, op):
    family = op["family"]
    wanted = set(op["keys"])
    if family == "accuracy_declarations":
        for name in list(pack["accuracy"]["declarations"]):
            if name in wanted:
                del pack["accuracy"]["declarations"][name]
        return {"keys": [f"declaration:{n}" for n in sorted(wanted)], "source": None,
                "source_digest": None, "authority": None}
    path, field = _DELETE_TARGETS[family]
    container, key = _walk(pack, path)
    container[key] = [r for r in container[key] if r[field] not in wanted]
    return {"keys": [f"{family}:{k}" for k in sorted(map(str, wanted))], "source": None,
            "source_digest": None, "authority": None}


def _op_set_check_param(pack, suite, op):
    suite[op["check"]]["params"][op["param"]] = op["value"]
    return {"keys": [f"param:{op['check']}.{op['param']}"], "source": None,
            "source_digest": None, "authority": None}


def _op_disable_check(pack, suite, op):
    suite[op["check"]]["enabled"] = False
    return {"keys": [f"check:{op['check']}"], "source": None, "source_digest": None,
            "authority": None}


def _op_rechain_log(pack, suite, op):
    prev = GENESIS_HASH
    for seq, entry in enumerate(sorted(pack["log"], key=lambda e: e["seq"])):
        entry["seq"] = seq
        entry["prev_hash"] = prev
        entry["entry_hash"] = digest_of({"seq": seq, "prev_hash": prev,
                                         "event": entry["event"]})
        prev = entry["entry_hash"]
    return {"keys": ["log:rechained"], "source": None, "source_digest": None,
            "authority": None}


def _op_set_log_anchor(pack, suite, op):
    head = op.get("head", "current")
    pack["log_anchor"] = (pack["log"][-1]["entry_hash"] if head == "current" else str(head))
    return {"keys": ["log:anchor"], "source": None, "source_digest": None, "authority": None}


def _op_edit_ledger_entry(pack, suite, op):
    entry = pack["ledger"][int(op["seq"])]
    entry["payload"][op["field"]] = op["value"]
    return {"keys": [f"ledger:{entry['seq']}"], "source": None, "source_digest": None,
            "authority": None}


def _op_set_field(pack, suite, op):
    container, key = _walk(pack, op["path"])
    container[key] = op["value"]
    return {"keys": [f"field:{op['path']}"], "source": None, "source_digest": None,
            "authority": None}


OPERATIONS = {
    # restore, recompute, re-seal: each one records where the new value came from
    "restore_log_segment": _op_restore_log_segment,
    "recompute_declaration": _op_recompute_declaration,
    "resource_section": _op_resource_section,
    "void_verifications": _op_void_verifications,
    "withhold_decisions": _op_withhold_decisions,
    "reassess_conformity": _op_reassess_conformity,
    "withdraw_declaration": _op_withdraw_declaration,
    "reissue_declaration": _op_reissue_declaration,
    "refit_limits_to_sealed_baseline": _op_refit_limits_to_sealed_baseline,
    "file_incident": _op_file_incident,
    # and the seven that change the check instead of the world
    "delete_records": _op_delete_records,
    "set_check_param": _op_set_check_param,
    "disable_check": _op_disable_check,
    "rechain_log": _op_rechain_log,
    "set_log_anchor": _op_set_log_anchor,
    "edit_ledger_entry": _op_edit_ledger_entry,
    "set_field": _op_set_field,
}

HONEST_OPS = ("restore_log_segment", "recompute_declaration", "resource_section",
              "void_verifications", "withhold_decisions", "reassess_conformity",
              "withdraw_declaration", "reissue_declaration",
              "refit_limits_to_sealed_baseline", "file_incident")


def apply_repairs(pack: dict, suite: dict, plan: list) -> tuple:
    """Apply a repair plan to copies of the pack and the suite. Given to you; not graded.

    Every operation appends one entry to `pack["repair_ledger"]` saying which evidence keys
    it touched, which source the new value came from — `None` for the seven operations that
    have no source — and who authorised it.
    """
    pack, suite = copy.deepcopy(pack), copy.deepcopy(suite)
    for number, op in enumerate(plan):
        name = op.get("op")
        if name not in OPERATIONS:
            raise ValueError(f"operation {number} is {name!r}; the vocabulary is "
                             f"{sorted(OPERATIONS)}")
        record = OPERATIONS[name](pack, suite, op)
        pack["repair_ledger"].append({"n": number, "op": name, "finding": op.get("finding"),
                                      **record})
    return pack, suite


print(f"{len(OPERATIONS)} operations · {len(HONEST_OPS)} that restore, recompute or re-seal "
      f"· {len(OPERATIONS) - len(HONEST_OPS)} that do not")
_index = evidence_index(PACK)
print(f"evidence index: {sum(len(v) for v in _index.values())} keys across "
      f"{len(_index)} families · " +
      " ".join(f"{k}={len(v)}" for k, v in sorted(_index.items())))
```

## 6. Exercise 1 — run the suite

The first thing an inspector needs is not a verdict, it is a table. Six checks, six answers,
and one distinction that decides everything downstream: a check that **ran and passed** is
not the same as a check **nobody ran**.

Every silenced pack you will ever meet is green. Some of them are green because a check was
switched off and the runner reported the absence of a failure as a pass. So `run_suite` has
three statuses, not two, and `verdict` is `"pass"` only when every check ran *and* every
check was green.

<details><summary>💡 Hint 1 — what to think about</summary>

Before you call anything, ask two questions of each check id: is it switched on, and
does anything implement it? A check you never called has no counts at all, and a 0 would
read as "we looked and found nothing". Then ask what the verdict should be for a suite
with every check switched off.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Walk the suite's ids in sorted order. Settle the not-run cases first — switched off,
then unknown to `checks` — each with its own documented reason code and empty fields, and
without calling anything. Only then call the check with its own `params`, turn its `green`
flag into a status, and copy the other four fields across untouched. Build the three lists
from the statuses, and let the verdict pass only when the red list and the not-run list
are both empty.
</details>

```
def run_suite(pack: dict, suite: dict, checks: dict = CHECKS) -> dict:
    """Run every check the suite lists, and report what each one says.

    For each check id in `suite`, in sorted order:

    * `suite[cid]["enabled"]` falsy -> status "not_run", reason_code "not_verified", and
      nothing else: `affected` None, `population` None, `detail` {}. Do not call the check.
    * an id with no entry in `checks` -> status "not_run", reason_code "no_such_check", the
      same empty fields. A suite that names a check nobody implemented is not a passing
      suite.
    * otherwise call `checks[cid](pack, suite[cid]["params"])` and carry its answer through:
      status "green" when `result["green"]` else "red", plus its `reason_code`, `affected`,
      `population` and `detail` unchanged.

    Returns a dict with exactly these keys:
      "results"  {check_id: {"status", "reason_code", "affected", "population", "detail"}}
      "green"    sorted list of the ids whose status is "green"
      "red"      sorted list of the ids whose status is "red"
      "not_run"  sorted list of the ids whose status is "not_run"
      "verdict"  "pass" only when "red" and "not_run" are BOTH empty, else "fail"

    Example:
        >>> run = run_suite(PACK, SUITE)
        >>> run["verdict"], len(run["red"]), run["not_run"]
        ('fail', 6, [])
        >>> run["results"]["log_integrity"]["reason_code"]
        'purged_window'
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_run_suite() -> None:
    run = run_suite(PACK, SUITE)
    assert set(run) == {"results", "green", "red", "not_run", "verdict"}, (
        f"run_suite returned {sorted(run)} — return exactly the five documented keys")
    assert len(run["results"]) == len(SUITE), (
        f"{len(run['results'])} results for {len(SUITE)} checks")
    assert run["red"] == sorted(CHECKS), (
        f"red came back {run['red']} — every one of the six checks fails on this pack, and "
        "the provider's attestation says all twelve of theirs passed")
    assert run["verdict"] == "fail" and run["not_run"] == [], (
        f"verdict {run['verdict']!r} with not_run {run['not_run']} — every check here is "
        "enabled and implemented, so nothing is unverified, and six are red, so 'fail'")
    off = copy.deepcopy(SUITE)
    off["oversight_plausibility"]["enabled"] = False
    quiet = run_suite(PACK, off)
    assert quiet["not_run"] == ["oversight_plausibility"], (
        "a disabled check must come back 'not_run'; it did not")
    assert "oversight_plausibility" not in quiet["green"], (
        "you reported a disabled check as green. That is the whole lesson: nobody looked, "
        "so nobody can say it passed")
    assert quiet["results"]["oversight_plausibility"]["affected"] is None, (
        "a check that did not run has no affected count — leave it None rather than 0, "
        "which reads as 'we looked and found nothing'")
    every_off = {cid: {**row, "enabled": False} for cid, row in SUITE.items()}
    assert run_suite(PACK, every_off)["verdict"] == "fail", (
        "a suite with every check switched off came back 'pass'. Verdict is 'pass' only "
        "when nothing is red AND nothing is unverified")
    print(f"suite: {len(run['red'])} red, {len(run['green'])} green, "
          f"{len(run['not_run'])} not run · verdict {run['verdict']}")
    for cid in run["red"]:
        row = run["results"][cid]
        print(f"  {cid:24s} {row['reason_code']:44s} {row['affected']}/{row['population']}")


_try("exercise 1", _check_run_suite)
```

## 7. Exercise 2 — the findings, worst first

Six red rows is not a finding list. A finding says which evidence id is in doubt, which
article it hangs off, which module's tool caught it, **how much** of that evidence the
defect touches, and exactly which keys in the evidence index it is about — because section
11 will ask whether the repair stayed inside them.

Two rules with teeth.

* A `not_run` check is a finding too. Not a smaller one: the share of the evidence it
  leaves in doubt is **all of it**, so it sorts to the top. A pack whose oversight check
  was switched off is in a worse state than one whose oversight check found four hundred
  bad reviews, because at least the second one has a number.
* Rank by share, not by count. Fourteen purged log entries out of 441 and one undeclarable
  metric out of three are not the same size of hole, and sorting by `affected` says they
  are.

<details><summary>💡 Hint 1 — what to think about</summary>

Which rows become findings? Everything that is not green — and the not-run rows are the
ones people drop. Then ask what "how big" means: a few purged log entries out of hundreds
sounds larger than one bad metric until you divide each by what it came from. A not-run
row has nothing to divide, so its share has to come from what not running means.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Skip green rows and any id `CHECK_META` does not know. Copy evidence id, module and
basis out of `CHECK_META`, and the reason code, counts and detail out of the result,
unchanged; only the share is derived — fixed for a check nobody ran, affected over
population for a red one, and zero whenever that division cannot be done. Sort the
detail's keys, treating an absent list as empty. Sort the findings on one key, share
largest first and then finding id alphabetically: reversing the whole sort would reverse
the tie-break too.
</details>

```
def inspect_pack(run: dict) -> list:
    """Turn a suite run into findings, worst first.

    One finding for every check whose status is NOT "green" — the red ones and the
    `not_run` ones alike. Each finding is a dict with exactly these keys:

      "finding_id"    f"{check_id}:{reason_code}"
      "check"         the check id
      "evidence_id"   CHECK_META[check]["evidence_id"]
      "module"        CHECK_META[check]["module"]
      "basis"         CHECK_META[check]["basis"]
      "reason_code"   the result's reason code
      "affected"      the result's affected count, unchanged (None for not_run)
      "population"    the result's population, unchanged (None for not_run)
      "share"         affected / population for a red check — 0.0 when population is 0 or
                      either number is None. For a not_run check it is 1.0: nobody looked
                      at any of it, so none of it is verified.
      "evidence_keys" sorted list of `detail["keys"]`, or [] when the detail has none
      "detail"        the result's detail, unchanged

    A check id the suite ran that `CHECK_META` does not know is skipped, not crashed on.

    Sort by "share" DESCENDING, then by "finding_id" ascending.

    Returns: list of finding dicts.

    Example:
        >>> findings = inspect_pack(run_suite(PACK, SUITE))
        >>> len(findings), findings[0]["check"]
        (6, 'declared_accuracy')
        >>> findings[-1]["finding_id"]
        'log_integrity:purged_window'
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_inspect_pack() -> None:
    findings = inspect_pack(run_suite(PACK, SUITE))
    assert len(findings) == 6, f"{len(findings)} findings, expected 6"
    order = [f["check"] for f in findings]
    assert order == ["declared_accuracy", "oversight_plausibility", "monitoring_limits",
                     "declaration_currency", "document_sourcing", "log_integrity"], (
        f"findings came back in the order {order}. Rank by share of the evidence affected, "
        "descending — 1 of 3 declared metrics is a bigger hole than 14 of 441 log entries")
    assert abs(findings[0]["share"] - 1 / 3) < 1e-12, (
        f"the worst finding's share is {findings[0]['share']}, expected affected/population")
    assert findings[0]["evidence_id"] == "accuracy_robustness_cybersecurity_report", (
        f"the worst finding names evidence {findings[0]['evidence_id']!r}; evidence_id, "
        "module and basis are copied straight out of CHECK_META[check], not invented")
    keys = {f["check"]: f["evidence_keys"] for f in findings}
    assert keys["oversight_plausibility"][:1] == ["decision:DEC-0200"], (
        "evidence_keys must be the check's own detail['keys'], sorted — section 11 asks "
        "whether the repair stayed inside them")
    off = copy.deepcopy(SUITE)
    off["log_integrity"]["enabled"] = False
    quiet = inspect_pack(run_suite(PACK, off))
    assert quiet[0]["check"] == "log_integrity" and quiet[0]["share"] == 1.0, (
        "a check nobody ran must be a finding with share 1.0, at the top of the list: "
        f"got {quiet[0]['check']} with share {quiet[0]['share']}")
    assert quiet[0]["reason_code"] == "not_verified" and quiet[0]["affected"] is None, (
        f"a check nobody ran came back {quiet[0]['reason_code']!r} with affected "
        f"{quiet[0]['affected']!r} — pass reason_code and affected through unchanged; only "
        "'share' is derived, and 0 affected reads as 'we looked and found nothing'")
    assert len(quiet) == 6, (
        f"{len(quiet)} findings when one check is off, expected 6 — switching a check off "
        "does not remove its finding, it replaces it with a worse one")
    print(f"{len(findings)} findings · worst {findings[0]['finding_id']} at "
          f"{findings[0]['share']:.1%} of its evidence")
    for finding in findings:
        print(f"  {finding['share']:6.1%}  {finding['module']}  {finding['finding_id']}")


_try("exercise 2", _check_inspect_pack, needs=_UPSTREAM["exercise 2"])
```

## 8. Exercise 3 — what the repair did to the evidence

Now the half of the lesson that is not in any of modules 1 to 8.

A repair is a function from a pack to a pack, and the first question about one is not "is
it green now". It is: **what happened to the evidence?** Three answers matter — what was
added, what was changed, and what is no longer there — and only the third one is hard to
argue with. A check that was red and is now green over a pack with four hundred fewer
records in it has not been repaired.

`conserved` deliberately says less than it looks like it says. Read its docstring.

<details><summary>💡 Hint 1 — what to think about</summary>

A repair can drop a whole family, or add one the pack never had. If you loop over one
index's families, what happens to a family only the other index has? Then reread what
`conserved` promises, and decide which of added, mutated and removed it is allowed to
look at.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Take the union of the two indices' family names, and treat a family missing from either
side as empty there. Within a family, set differences of the key sets give removed and
added; mutated is the shared keys whose digests disagree. `n_before` and `n_after` are the
sizes of the two dicts, not of any difference. Sum each kind across the families for the
totals, and set `conserved` from the removed total alone.
</details>

```
def evidence_delta(before: dict, after: dict) -> dict:
    """Compare two evidence indices, family by family.

    Walk the UNION of both indices' families, so a repair that drops a whole family reports
    every one of its keys as removed rather than raising KeyError.

    For each family, a dict with exactly:
      "removed"   sorted keys present in `before` and absent from `after`
      "added"     sorted keys present in `after` and absent from `before`
      "mutated"   sorted keys in both whose digest differs
      "n_before"  how many keys that family had
      "n_after"   how many it has

    Plus two keys beside the families, in a dict with exactly these three top-level keys:
      "families"  the per-family dict above
      "totals"    {"removed", "added", "mutated"}, summed over every family
      "conserved" True when NOTHING was removed, anywhere.

    What "conserved" does not mean: a repair that rewrote every digest in the pack and
    removed nothing is still "conserved". Deleting the record is one way to silence a check,
    and this function measures exactly that one.

    Returns: dict with keys "families", "totals", "conserved".

    Example:
        >>> a = {"log": {"x": "1", "y": "2"}}
        >>> b = {"log": {"y": "3", "z": "4"}}
        >>> d = evidence_delta(a, b)
        >>> d["families"]["log"]["removed"], d["families"]["log"]["mutated"]
        (['x'], ['y'])
        >>> d["totals"], d["conserved"]
        ({'removed': 1, 'added': 1, 'mutated': 1}, False)
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_evidence_delta() -> None:
    same = evidence_delta(evidence_index(PACK), evidence_index(PACK))
    assert same["totals"] == {"removed": 0, "added": 0, "mutated": 0}, (
        f"a pack against itself came back {same['totals']} — the index is deterministic, so "
        "the only way to see a difference here is to compare the wrong things")
    assert same["conserved"] is True, (
        "a pack compared against itself came back not conserved — 'conserved' is about "
        "removals alone, and nothing was removed here")
    gone, _ = apply_repairs(PACK, SUITE, [
        {"op": "delete_records", "family": "oversight_verifications",
         "keys": [f"V-{i:04d}" for i in range(200, 600)]}])
    delta = evidence_delta(evidence_index(PACK), evidence_index(gone))
    assert set(delta) == {"families", "totals", "conserved"}, (
        f"evidence_delta returned {sorted(delta)} — three top-level keys, no more")
    assert delta["totals"]["removed"] == 400 and delta["conserved"] is False, (
        f"deleting 400 verifications showed {delta['totals']['removed']} removals and "
        f"conserved={delta['conserved']} — every deleted record is a removal, and a single "
        "removal anywhere means the delta is not conserved")
    assert delta["families"]["oversight"]["n_before"] == 2800, (
        "n_before is the size of the family BEFORE, not the size of the difference")
    assert delta["families"]["log"]["removed"] == [], "nothing was removed from the log"
    dropped = {"log": {}, "annex_iv": evidence_index(PACK)["annex_iv"]}
    partial = evidence_delta(evidence_index(PACK), dropped)
    assert partial["families"]["oversight"]["removed"] == sorted(
        evidence_index(PACK)["oversight"]), (
        "a family missing from the second index must report every key as removed, not raise")
    print(f"delta after deleting 400 records: {delta['totals']} · "
          f"conserved={delta['conserved']}")


_try("exercise 3", _check_evidence_delta)
```

## 9. Exercise 4 — what the repair did to the checks

The second way to make a red check green is to change the check. Switch it off, drop it
from the suite, or move the number it compares against. All three leave the evidence
untouched, so exercise 3 sees nothing at all.

The distinction that makes this exercise more than a `dict` diff: **tightening is not
silencing**. A repair that lowers `max_reviews_per_hour` from 60 to 20 has made the check
stricter, and an audit that flags it has cried wolf. `PARAM_DIRECTION` is what tells the
two apart, and a parameter it does not know about goes in `unknown` rather than being
guessed at in either direction.

<details><summary>💡 Hint 1 — what to think about</summary>

Every change has to answer one question before it is filed: does the check now let more
through? `PARAM_DIRECTION` answers it per parameter, and a parameter it has no entry for is
a question you cannot answer, so do not. Two changes are not moves of a number at all — a
parameter that vanished from `after`, and one that only `after` has. Which is a widening,
and which is a rule nobody has run yet?
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Check ids first: in `before` but not in `after` is a removed check, not a disabled one;
disabled means present in both and switched off only in `after`. For each check in both,
walk the parameters `before` had and skip any whose value did not change. Look the
direction up first — no entry means `unknown`, whatever happened to the parameter.
Otherwise a vanished parameter is loosened with `after` None, a move the weak way is
loosened and any other move is tightened. Ignore parameters only `after` has, sort the rows
by check then parameter, and compute `unchanged` from the four weakening lists alone.
</details>

```
def suite_delta(before: dict, after: dict, directions: dict = PARAM_DIRECTION) -> dict:
    """Compare two suites and report which changes WEAKEN the suite.

    Returns a dict with exactly these keys:
      "disabled"       sorted ids enabled in `before` and present but not enabled in `after`
      "enabled"        sorted ids not enabled in `before` and enabled in `after`
      "removed_checks" sorted ids in `before` and not in `after`
      "added_checks"   sorted ids in `after` and not in `before`
      "loosened"       rows {"check", "param", "before", "after"} for every parameter that
                       moved in the weakening direction: for direction "lower" that is
                       after < before, for "higher" it is after > before. A parameter that
                       `before` had and `after` does not is loosened too, with "after" None
                       — a limit you stopped applying is not a limit.
      "tightened"      the same rows for movement in the other direction. Never both.
      "unknown"        sorted f"{check}.{param}" for every parameter that changed and has no
                       entry in `directions`. Guessing a direction is worse than saying so.
      "unchanged"      True when "disabled", "removed_checks", "loosened" and "unknown" are
                       all empty. A suite that was tightened is still "unchanged" here —
                       this flag is about weakening, not about editing.

    A parameter that only `after` has is not a loosening and is not unknown: it is a new
    rule, and nothing that tests it has run yet. Ignore it.
    Rows in "loosened" and "tightened" are sorted by (check, param).

    Returns: dict with the eight keys above.

    Example:
        >>> wide = copy.deepcopy(SUITE)
        >>> wide["monitoring_limits"]["params"]["k"] = 5.0
        >>> suite_delta(SUITE, wide)["loosened"]
        [{'check': 'monitoring_limits', 'param': 'k', 'before': 3.0, 'after': 5.0}]
        >>> suite_delta(SUITE, wide)["unchanged"]
        False
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_suite_delta() -> None:
    assert suite_delta(SUITE, SUITE)["unchanged"] is True, (
        "a suite against itself came back changed")
    _, wide = apply_repairs(PACK, SUITE, [
        {"op": "set_check_param", "check": "oversight_plausibility",
         "param": "max_reviews_per_hour", "value": 1000},
        {"op": "set_check_param", "check": "oversight_plausibility",
         "param": "min_seconds", "value": 1}])
    got = suite_delta(SUITE, wide)
    assert set(got) == {"disabled", "enabled", "removed_checks", "added_checks", "loosened",
                        "tightened", "unknown", "unchanged"}, f"keys: {sorted(got)}"
    assert [r["param"] for r in got["loosened"]] == ["max_reviews_per_hour", "min_seconds"], (
        f"loosened came back {got['loosened']} — both parameters moved the weak way: "
        "max_reviews_per_hour is looser HIGHER and min_seconds is looser LOWER, and rows "
        "are sorted by (check, param)")
    assert got["tightened"] == [] and got["unchanged"] is False, (
        f"tightened came back {got['tightened']} and unchanged {got['unchanged']} — both "
        "parameters moved the WEAK way, so nothing is tightened and the suite is not "
        "unchanged")
    _, strict = apply_repairs(PACK, SUITE, [
        {"op": "set_check_param", "check": "oversight_plausibility",
         "param": "max_reviews_per_hour", "value": 20}])
    tighter = suite_delta(SUITE, strict)
    assert tighter["loosened"] == [] and len(tighter["tightened"]) == 1, (
        "lowering max_reviews_per_hour from 60 to 20 is a TIGHTER check; flagging it as "
        "loosened is how an audit teaches people to ignore it")
    assert tighter["unchanged"] is True, (
        "'unchanged' is about weakening: a suite that was only tightened must come back "
        "unchanged, or every tightening looks like an incident")
    _, off = apply_repairs(PACK, SUITE, [
        {"op": "disable_check", "check": "log_integrity"}])
    assert suite_delta(SUITE, off)["disabled"] == ["log_integrity"], (
        "a check enabled in 'before' and present-but-off in 'after' belongs in 'disabled'; "
        "switching a check off is the cheapest silencing there is")
    gone = {c: v for c, v in SUITE.items() if c != "monitoring_limits"}
    assert suite_delta(SUITE, gone)["removed_checks"] == ["monitoring_limits"], (
        "a check id in 'before' and absent from 'after' belongs in 'removed_checks', not "
        "in 'disabled' — walk both id sets, not only the ones they share")
    dropped_param = copy.deepcopy(SUITE)
    del dropped_param["monitoring_limits"]["params"]["k"]
    rows = suite_delta(SUITE, dropped_param)["loosened"]
    assert rows == [{"check": "monitoring_limits", "param": "k", "before": 3.0,
                     "after": None}], (
        f"a deleted parameter came back as {rows} — dropping the limit is the widest "
        "possible widening, and it must not be silently skipped")
    mystery = copy.deepcopy(SUITE)
    mystery["log_integrity"]["params"]["fudge"] = 1
    mystery2 = copy.deepcopy(mystery)
    mystery2["log_integrity"]["params"]["fudge"] = 99
    assert suite_delta(mystery, mystery2)["unknown"] == ["log_integrity.fudge"], (
        "a parameter PARAM_DIRECTION says nothing about goes in 'unknown', spelled "
        "f'{check}.{param}' — guessing a direction is worse than saying you do not know")
    print(f"widening two thresholds: {len(got['loosened'])} loosened, "
          f"{len(got['tightened'])} tightened, unchanged={got['unchanged']}")


_try("exercise 4", _check_suite_delta)
```

## 10. Exercise 5 — the numbers that cannot be reproduced

The third way to make a red check green is to write the answer in. Nothing is deleted,
nothing is switched off, and a metric with no interval acquires one that looks exactly
like a measurement.

The defence is the one module 6 built: **retain the evidence and recompute**. Every number
this pack declares has something behind it — the evaluation split behind the interval, the
artefact behind the document section, the sealed baseline behind the control limits, and
the hash chain behind all three. A value that does not come back when you recompute it is
not evidence, whatever the repair ledger says about where it came from. This function
never reads the repair ledger, on purpose.

<details><summary>💡 Hint 1 — what to think about</summary>

Every rule is the same move: take the number the pack states, find the evidence it says
it came from, compute it again, and compare. The mistakes are borrowing a setting from
somewhere else — the suite's `k` instead of the one the limits record, a default seed
instead of the declaration's own, any declaration instead of the one the artefact names —
and comparing one field when the rule lists several.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

One pass per rule, each adding at most one row per claim: stop at the first reason that
applies. Declarations: the required keys, then the seed, then recompute over the retained
evaluation split with the metric, resample count, level and seed stored in the
declaration, and compare point, lo and hi within `tol`. Sections: the emitted value against
the field it cites, then its stored digest against the artefact's payload; separately, a
sourced field's `_source` stamp against the digest of the declaration for that artefact's
own `primary_metric`. Limits: rebuild from the sealed baseline at the limits' own `k` and
compare all four numbers. Ledger: `verify_ledger`. Sort by claim, and never open the
repair ledger.
</details>

```
CLAIM_CHECKS = (("accuracy.declarations", ("declared_accuracy",)),
                ("annex_iv.sections", ("document_sourcing",)),
                ("sources.", ("document_sourcing",)),
                ("monitoring.limits", ("monitoring_limits",)),
                ("ledger", ("monitoring_limits", "declaration_currency")))


def checks_for_claim(claim: str) -> tuple:
    """Which checks a failed claim undermines. Given to you; not graded."""
    for prefix, checks in CLAIM_CHECKS:
        if claim.startswith(prefix):
            return checks
    return ()


def unsupported_values(pack: dict, tol: float = 1e-9) -> list:
    """Every claim in the pack that does not reproduce from the evidence it cites.

    Four rules, each of which RECOMPUTES something.

    1. For each name in `pack["accuracy"]["declarations"]`, claim
       f"accuracy.declarations.{name}":
         * any of DECLARATION_KEYS absent          -> reason "no_interval"
         * "seed" absent                           -> reason "no_recompute_spec"
         * otherwise recompute with bootstrap_ci over pack["accuracy"]["evaluation"],
           `metric=METRICS[declaration["metric"]]`, and the declaration's own "n_boot",
           "level" and "seed"; if "point", "lo" or "hi" differ by more than `tol`
                                                   -> reason "does_not_reproduce"
    2. For each emitted Annex IV section carrying a "value", claim
       f"annex_iv.sections.{number}":
         * value != field_value(pack["sources"], section["source_path"])
                                                   -> "does_not_match_its_source"
         * section["source_digest"] != digest_of(that artefact's payload)
                                                   -> "stale_source_digest"
       and, when that field IS sourced, claim f"sources.{artefact}.{field}":
         * the artefact payload's f"{field}_source" is not the digest of the declaration the
           artefact itself names in "primary_metric". The artefact says which metric it is
           about; a number stamped with a different metric's digest is that other metric's
           number, sitting in this one's row  -> "not_traceable_to_a_measurement"
    3. Claim "monitoring.limits": recompute control_limits() from the sealed baseline
       ledger entry, at the k the limits themselves record.
         * no sealed baseline entry                -> "no_sealed_baseline"
         * any of centre, sigma, ucl, lcl differs by more than `tol`
                                                   -> "not_from_the_sealed_baseline"
    4. Claim "ledger": verify_ledger(pack["ledger"]) -> "ledger_chain_broken" when it is not
       intact.

    Report at most ONE reason per claim: the first that applies, in the order above.

    Returns: a list of {"claim", "reason", "expected", "found"}, sorted by "claim".
    `expected` and `found` are whatever makes the difference readable; both may be None.

    Example:
        >>> [u["claim"] for u in unsupported_values(PACK)]
        ['accuracy.declarations.accuracy', 'monitoring.limits']
        >>> unsupported_values(PACK)[0]["reason"]
        'no_interval'
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_unsupported_values() -> None:
    shipped = unsupported_values(PACK)
    assert [u["claim"] for u in shipped] == ["accuracy.declarations.accuracy",
                                             "monitoring.limits"], (
        f"the shipped pack came back with {[u['claim'] for u in shipped]}, expected the "
        "bare accuracy point estimate and the refitted control limits")
    assert shipped[0]["reason"] == "no_interval", (
        f"the accuracy declaration came back {shipped[0]['reason']}: it has no lo, no hi, "
        "no level, no n, no n_boot and no method, so there is nothing to recompute yet")
    assert shipped[1]["reason"] == "not_from_the_sealed_baseline", (
        f"the control limits came back {shipped[1]['reason']}: recompute centre, sigma, ucl "
        "AND lcl with control_limits() from the sealed baseline entry, at the k the limits "
        "themselves record, and compare all four")
    faked, _ = apply_repairs(PACK, SUITE, [
        {"op": "set_field", "path": "accuracy.declarations.accuracy",
         "value": {"name": "accuracy", "metric": "accuracy", "point": 0.9, "lo": 0.89,
                   "hi": 0.91, "level": 0.95, "n": 1200, "n_boot": 400,
                   "method": "percentile bootstrap", "seed": 90210}}])
    claims = {u["claim"]: u["reason"] for u in unsupported_values(faked)}
    assert claims.get("accuracy.declarations.accuracy") == "does_not_reproduce", (
        "an interval typed in by hand passed. Recompute it from "
        "pack['accuracy']['evaluation'] with the declaration's own seed and n_boot and "
        "compare all three of point, lo and hi")
    edited, _ = apply_repairs(PACK, SUITE, [
        {"op": "edit_ledger_entry", "seq": 1, "field": "errors", "value": 132}])
    claims = {u["claim"]: u["reason"] for u in unsupported_values(edited)}
    assert claims.get("ledger") == "ledger_chain_broken", (
        "editing a sealed ledger payload went unnoticed — recompute each entry's digest "
        "over seq, kind, payload and prev, the way ledger_append computed it")
    hand, _ = apply_repairs(PACK, SUITE, [
        {"op": "set_field", "path": "sources.metrics.payload.primary_metric_value",
         "value": 0.9}])
    claims = {u["claim"]: u["reason"] for u in unsupported_values(hand)}
    assert claims.get("annex_iv.sections.4") == "does_not_match_its_source", (
        "writing a number into the artefact while the emitted section still shows the old "
        "one must be caught: the section and its source no longer agree")
    assert claims.get("sources.metrics.primary_metric_value") == \
        "not_traceable_to_a_measurement", (
        "a metric value with no digest linking it to the declaration of the metric the "
        "artefact says it is about came from somebody's head. Check f'{field}_source' "
        "against the digest of declarations[payload['primary_metric']]")
    print("unsupported on the shipped pack: " +
          ", ".join(f"{u['claim']} ({u['reason']})" for u in shipped))


_try("exercise 5", _check_unsupported_values)
```

## 11. Exercise 6 — repaired, silenced, or unfixed

This is the exercise the capstone exists for. You have three measurements of what a repair
did — to the evidence, to the checks, and to the numbers — and one question to answer with
them, once per finding:

* **unfixed** — the check is still red. Honest, incomplete, and the easiest of the three to
  live with.
* **silenced** — the check is green, and it is green for a reason that has nothing to do
  with the system: the check did not run, a threshold moved, the records went away, the
  repair rewrote evidence the finding never mentioned, or a number in the pack no longer
  reproduces.
* **repaired** — the check is green and none of the above is true.

The order matters. A check that did not run is never "unfixed" and never "repaired": it is
the cleanest silencing there is, and reporting it as anything else is the failure this
lesson is about.

<details><summary>💡 Hint 1 — what to think about</summary>

The verdict is a ladder, and the order of its rungs is the lesson. A check that did not
run can never be unfixed or repaired; a check that is still red is unfixed whatever else
the repair did. Only a green check earns the four questions, and each one is about this
check alone: a widened threshold on another check, a removal from another family, or a
rewrite of a key the finding itself named is not evidence against this one.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

For each finding, look its check up in the after-run. Missing or not run: silenced, plus
the removed-from-suite reason when the suite change lists it. Red: unfixed, with the reason
code the after-run reports now, not the one in the finding. Green: build the reasons in the
docstring's four steps, finishing each step across all the check's families before
starting the next, leaving out mutated keys the finding named, and skipping a family the
delta does not have. Count all three verdicts, zeros included. Collateral is green before
and anything else after, not-run included; honest needs every finding repaired AND no
collateral.
</details>

```
def silencing_audit(findings: list, before_run: dict, after_run: dict, delta: dict,
                    suite_change: dict, unsupported: list) -> dict:
    """Decide, for each finding, whether the repair fixed it, silenced it, or left it.

    For each finding in `findings`, in the order given, look at
    `after_run["results"].get(finding["check"])`:

    * absent, or status "not_run"  -> verdict "silenced", reasons ["check_not_run"], plus
      "check_removed_from_suite" when the check is in `suite_change["removed_checks"]`.
    * status "red"                 -> verdict "unfixed", reasons
      [f"still_red:{reason_code}"].
    * status "green"               -> collect reasons in exactly this order, and the verdict
      is "silenced" when there is at least one and "repaired" when there is none:
        1. f"param_loosened:{param}" for each row of `suite_change["loosened"]` whose
           "check" is this check, in the order they appear.
        2. f"evidence_removed:{family}:{n}" for each family in CHECK_FAMILIES[check], in
           that tuple's order, where n = len(delta["families"][family]["removed"]) and
           n > 0.
        3. f"evidence_changed_outside_the_finding:{family}:{n}" for each family in the same
           order, where n counts the family's MUTATED keys that are not in the finding's
           "evidence_keys" — the repair changed evidence this finding never named.
        4. f"unsupported:{claim}" for each row of `unsupported`, in the order given, whose
           `checks_for_claim(claim)` includes this check.
      A family missing from `delta["families"]` contributes nothing and must not raise.

    Returns a dict with exactly:
      "verdicts"   {finding_id: {"verdict", "reasons"}}
      "counts"     {"repaired", "silenced", "unfixed"}
      "collateral" sorted check ids that were "green" in `before_run` and are not "green"
                   in `after_run` — a repair that broke something nobody complained about
      "honest"     True only when every verdict is "repaired" AND "collateral" is empty

    Example:
        >>> res = audit_repair(PACK, SUITE, TEMPTING_PLANS["stop running the check"])
        >>> res["audit"]["verdicts"][
        ...     "oversight_plausibility:implausible_review"]["verdict"]
        'silenced'
    """
    # YOUR CODE HERE
    raise NotImplementedError


def audit_repair(pack: dict, suite: dict, plan: list) -> dict:
    """The whole pipeline in one call, on YOUR six functions. Given to you; not graded."""
    before_run = run_suite(pack, suite)
    findings = inspect_pack(before_run)
    after_pack, after_suite = apply_repairs(pack, suite, plan)
    after_run = run_suite(after_pack, after_suite)
    delta = evidence_delta(evidence_index(pack), evidence_index(after_pack))
    change = suite_delta(suite, after_suite)
    unsupported = unsupported_values(after_pack)
    return {"findings": findings, "before_run": before_run, "after_run": after_run,
            "after_pack": after_pack, "after_suite": after_suite, "delta": delta,
            "suite_change": change, "unsupported": unsupported,
            "audit": silencing_audit(findings, before_run, after_run, delta, change,
                                     unsupported)}


# Five repairs, each of which turns a red check green and fixes nothing. They are written in
# the same operations vocabulary as an honest repair, because in a real pack that is exactly
# how they arrive: as a pull request that makes the pipeline pass.
_STAMPED = [f"V-{i:04d}" for i in range(200, 200 + N_STAMPED)]
_STAMPED_DECISIONS = [f"DEC-{i:04d}" for i in range(200, 200 + N_STAMPED)]

TEMPTING_PLANS = {
    "delete the failing records": [
        {"finding": "oversight_plausibility:implausible_review", "op": "delete_records",
         "family": "oversight_verifications", "keys": _STAMPED},
        {"finding": "oversight_plausibility:implausible_review", "op": "delete_records",
         "family": "oversight_decisions", "keys": _STAMPED_DECISIONS}],
    "widen the tolerance": [
        {"finding": "oversight_plausibility:implausible_review", "op": "set_check_param",
         "check": "oversight_plausibility", "param": "min_seconds", "value": 1},
        {"finding": "oversight_plausibility:implausible_review", "op": "set_check_param",
         "check": "oversight_plausibility", "param": "max_reviews_per_hour",
         "value": 1000}],
    "stop running the check": [
        {"finding": "oversight_plausibility:implausible_review", "op": "disable_check",
         "check": "oversight_plausibility"}],
    "re-chain the log and republish the anchor": [
        {"finding": "log_integrity:purged_window", "op": "rechain_log"},
        {"finding": "log_integrity:purged_window", "op": "set_log_anchor",
         "head": "current"}],
    "type the interval in": [
        {"finding": "declared_accuracy:no_interval", "op": "set_field",
         "path": "accuracy.declarations.accuracy",
         "value": {"name": "accuracy", "metric": "accuracy", "point": 0.9, "lo": 0.88,
                   "hi": 0.92, "level": 0.95, "n": N_EVAL, "n_boot": N_BOOT,
                   "method": "percentile bootstrap", "seed": BOOT_SEED}}],
}

_TEMPTING_TARGET = {"delete the failing records": "oversight_plausibility",
                    "widen the tolerance": "oversight_plausibility",
                    "stop running the check": "oversight_plausibility",
                    "re-chain the log and republish the anchor": "log_integrity",
                    "type the interval in": "declared_accuracy"}


def _check_silencing_audit() -> None:
    for name, plan in TEMPTING_PLANS.items():
        result = audit_repair(PACK, SUITE, plan)
        check = _TEMPTING_TARGET[name]
        finding = next(f for f in result["findings"] if f["check"] == check)
        verdict = result["audit"]["verdicts"][finding["finding_id"]]
        status = result["after_run"]["results"].get(check, {}).get("status", "not_run")
        assert status in ("green", "not_run"), (
            f"{name!r} was supposed to turn {check} green and left it {status}")
        assert verdict["verdict"] == "silenced", (
            f"{name!r} left {check} {status} and your audit called the finding "
            f"{verdict['verdict']!r}. Every one of these five fixes nothing: the reasons "
            "are a check that did not run, a widened threshold, deleted records, rewritten "
            "evidence and a number that does not reproduce")
        assert verdict["reasons"], "a silenced verdict must say which signal fired"
    result = audit_repair(PACK, SUITE, [])
    counts = result["audit"]["counts"]
    assert counts == {"repaired": 0, "silenced": 0, "unfixed": 6}, (
        f"an empty repair plan scored {counts} — every finding is still red, and 'unfixed' "
        "is the honest word for that")
    assert result["audit"]["honest"] is False, (
        "'honest' came back True for a plan that repaired nothing — it is True only when "
        "every verdict is 'repaired' AND nothing was broken in passing")
    reasons = result["audit"]["verdicts"]["log_integrity:purged_window"]["reasons"]
    assert reasons == ["still_red:purged_window"], (
        f"an unfixed finding's reason came back {reasons}, expected the reason code the "
        "check itself still reports, prefixed with still_red:")
    assert result["audit"]["collateral"] == [], (
        f"collateral came back {result['audit']['collateral']} on a pack nobody touched — "
        "collateral is the checks that were green BEFORE and are not green after, and "
        "nothing was green before")
    print("all five tempting repairs: verdict silenced")


_try("exercise 6", _check_silencing_audit, needs=_UPSTREAM["exercise 6"])
```

## 13. Exercise 7 — the honest repair

Now do it properly. `plan_repairs` returns a list of operations, in the order they must be
applied, drawn only from `HONEST_OPS`. Everything the honest repair needs is in the pack
already: the purged entries are in cold storage, the evaluation split behind the metric was
retained, the baseline was sealed before the period it judges, and the changes the
declaration ignored are in the change log.

Two of the six take more than one operation, and both for the same reason: undoing the
artefact is not the repair. Voiding four hundred implausible reviews leaves four hundred
decisions standing on nothing, so the decisions have to be withheld too. Refitting the
control limits un-hides an excursion nobody has filed, so the incident has to be filed.
A repair that stops halfway turns one red check into a different red check, which is
**unfixed** — and that is still a better outcome than green.

<details><summary>💡 Hint 1 — what to think about</summary>

Every repair reads its target out of the finding's detail: the missing sequence numbers,
the empty path, the stamped ids, the latest substantial change, the hidden windows.
Remembering them from this notebook is the trap — the grader also runs your plan on a
different pack with the defects in different places. And two of the six repairs take two
steps: ask what the check will say after the first step alone.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Walk the findings in the order given, branch on the reason code, and tag every operation
with its finding id. Log: the cold-store export whose `covers` range overlaps the missing
seqs, not the first one listed. Accuracy: one recompute per metric named. Document: split
each empty path into artefact and field, and take the metric from that artefact's own
`primary_metric`. Oversight: strip the prefixes, void the verifications with an authority
and a reason, then withhold the decisions, with a reason too. Declaration: withdraw with a
reason, then reassess and reissue, each dated no earlier than what it has to cover. Limits:
refit, then file one incident, aware today, over the hidden windows. Any other code:
nothing.
</details>

```
def plan_repairs(pack: dict, findings: list) -> list:
    """Plan the honest repair: the operations, in the order they must be applied.

    Use only the ten operations in HONEST_OPS. Give every operation a "finding" key holding
    the finding_id it belongs to. Walk `findings` in the order given — it is already sorted
    worst first, and that order happens to put the accuracy repair before the document
    repair, which needs the number it produces.

    By `reason_code`:

    "purged_window" -> one "restore_log_segment" whose "segment" is the key in
        `pack["cold_store"]` whose "covers" range overlaps `detail["missing_seq"]`. There are
        two exports in cold storage; the other one holds entries the log already has, and
        restoring it changes nothing.

    "no_interval" -> one "recompute_declaration" per metric in `detail["metrics"]`, with
        "n_boot" and "seed". Any values reproduce, because the seed is stored with the
        declaration; N_BOOT and BOOT_SEED are the obvious pair.

    "unsourced_section" -> one "resource_section" per row of `detail["detail_rows"]`, with
        "section" the row's number, and "artefact" and "field" split out of the empty path
        it reports ("metrics.primary_metric_value" -> artefact "metrics", field
        "primary_metric_value"). "from_declaration" is the metric that artefact is about:
        `pack["sources"][artefact]["payload"]["primary_metric"]`.

    "implausible_review" -> "void_verifications" over every `verification:` key in the
        finding's evidence_keys (ids without the prefix), with a non-empty "authority" and
        "reason"; then "withhold_decisions" over every `decision:` key, with a non-empty
        "reason".

    "declaration_predates_substantial_modification" -> "withdraw_declaration" with a reason,
        then "reassess_conformity" dated no earlier than `detail["latest_substantial"]`,
        then "reissue_declaration" dated no earlier than that assessment. AS_OF works for
        both, and both operations refuse a date that would put the paperwork before the
        change it is supposed to cover.

    "limits_not_from_sealed_baseline" -> "refit_limits_to_sealed_baseline", then one
        "file_incident" with an "incident_id", an "aware_on" date — the Article 73 clock
        starts at awareness, which is today — and "windows" set to `detail["windows_hidden"]`,
        the windows the refitted limits were hiding.

    Any other reason_code: emit nothing for it rather than guessing.

    Returns: list of operation dicts.

    Example:
        >>> plan = plan_repairs(PACK, inspect_pack(run_suite(PACK, SUITE)))
        >>> [op["op"] for op in plan][:2]
        ['recompute_declaration', 'void_verifications']
        >>> all(op["op"] in HONEST_OPS for op in plan)
        True
    """
    # YOUR CODE HERE
    raise NotImplementedError


def _check_plan_repairs() -> None:
    findings = inspect_pack(run_suite(PACK, SUITE))
    plan = plan_repairs(PACK, findings)
    assert plan, "plan_repairs returned nothing"
    stray = sorted({op["op"] for op in plan} - set(HONEST_OPS))
    assert not stray, (
        f"the plan uses {stray}, which change the check rather than the world. The honest "
        f"vocabulary is {list(HONEST_OPS)}")
    assert all(op.get("finding") for op in plan), (
        "every operation needs a 'finding' key: a repair nobody can trace to a finding is "
        "an unexplained edit to an evidence pack")
    result = audit_repair(PACK, SUITE, plan)
    still_red = result["after_run"]["red"]
    assert not still_red, (
        f"after your repair these checks are still red: {still_red}. Read each one's "
        "reason_code — voiding without withholding, and refitting without filing, each "
        "turn one red into a different red")
    assert result["after_run"]["not_run"] == [], "an honest repair runs every check"
    audit = result["audit"]
    silenced = {f: v["reasons"] for f, v in audit["verdicts"].items()
                if v["verdict"] == "silenced"}
    assert not silenced, f"your own audit calls these silenced: {silenced}"
    assert audit["honest"] is True and audit["counts"]["repaired"] == 6, (
        f"your own audit scored the honest plan {audit['counts']} with "
        f"honest={audit['honest']} — all six findings come back 'repaired', and no check "
        "that was green is broken in passing")
    assert result["delta"]["conserved"] is True, (
        "the repair removed evidence: an honest repair to this pack adds, re-seals and "
        "supersedes, and deletes nothing")
    assert result["suite_change"]["unchanged"] is True, "the repair edited the suite"
    assert result["unsupported"] == [], (
        f"these numbers still do not reproduce after the repair: {result['unsupported']}")
    segments = {op["segment"] for op in plan if op["op"] == "restore_log_segment"}
    assert segments == {"export-2026-05-14"}, (
        f"restored {segments} — pick the export whose 'covers' range overlaps the missing "
        "sequence numbers; the other one restores entries the log already has")
    print(f"honest repair: {len(plan)} operations · "
          f"{len(result['after_run']['green'])} checks green · "
          f"audit {result['audit']['counts']} · evidence "
          f"+{result['delta']['totals']['added']} changed "
          f"{result['delta']['totals']['mutated']} removed "
          f"{result['delta']['totals']['removed']}")


_try("exercise 7", _check_plan_repairs, needs=_UPSTREAM["exercise 7"])
```

## 14. Exercise 8 — the inspection report

The artefact. One page, per finding, with the verdict and the reason beside it, and one
word at the bottom. The word is not "green".

The precedence is the argument. A pack with one silenced finding and five repaired ones is
in a worse state than a pack with six unfixed ones, because the second pack is honest about
where it is and the first one is not. So **silenced** outranks **unfixed**, and a repair
that broke a check nobody had complained about — `collateral` — is also a rejection.

<details><summary>💡 Hint 1 — what to think about</summary>

Nearly every field is copied, and the traps are in what gets copied: the checks the run
covered rather than the findings, the claim names rather than the claim rows, and the
findings in their own worst-first order rather than grouped by verdict. The one judgement
is the last word: ask why a pack that has been made green deserves a harsher one than a
pack that is honestly red.
</details>
<details><summary>💡 Hint 2 — the approach, in words</summary>

Take each field from the place the docstring names in the `audit_repair()` result. Build
each row from the four fields the docstring copies from the finding (which are not the
finding's first four keys) plus the audit's verdict and reasons for that finding id,
keeping the findings' order. Then decide the verdict top down: any silenced finding first;
then any unfixed finding or any collateral; only then accepted.
</details>

```
def inspection_report(result: dict, as_of: date = AS_OF) -> dict:
    """Assemble the inspection report from a completed audit_repair() result.

    Returns a dict with exactly these keys:
      "system_id"       result["after_pack"]["system_id"]
      "as_of"           as_of.isoformat()
      "checks"          how many checks the run covered: len(before_run["results"])
      "green_before"    len(result["before_run"]["green"])
      "green_after"     len(result["after_run"]["green"])
      "findings"        one row per finding, IN result["findings"] ORDER, each a dict with
                        exactly "finding_id", "evidence_id", "module", "share", "verdict"
                        and "reasons" — the first four copied from the finding, the last two
                        from result["audit"]["verdicts"][finding_id]
      "counts"          result["audit"]["counts"], unchanged
      "collateral"      result["audit"]["collateral"], unchanged
      "evidence"        result["delta"]["totals"], unchanged
      "suite_unchanged" result["suite_change"]["unchanged"]
      "unsupported"     sorted claims from result["unsupported"] that still do not reproduce
      "verdict"         "rejected: evidence silenced"    when counts["silenced"] > 0
                        "rejected: findings unresolved"  when counts["unfixed"] > 0 or
                                                         "collateral" is non-empty
                        "accepted"                       otherwise
                        in that order of precedence: a silenced finding outranks an
                        unresolved one, because the second is at least honest.

    Returns: the report dict.

    Example:
        >>> report = inspection_report(audit_repair(PACK, SUITE, []))
        >>> report["verdict"], report["counts"]["unfixed"]
        ('rejected: findings unresolved', 6)
    """
    # YOUR CODE HERE
    raise NotImplementedError


# A pack made green the other way: one silencing operation per finding, six for six.
SILENT_PLAN = (TEMPTING_PLANS["delete the failing records"]
               + TEMPTING_PLANS["re-chain the log and republish the anchor"]
               + TEMPTING_PLANS["type the interval in"]
               + [{"finding": "document_sourcing:unsourced_section", "op": "set_field",
                   "path": "sources.metrics.payload.primary_metric_value", "value": 0.9},
                  {"finding": "declaration_currency:"
                              "declaration_predates_substantial_modification",
                   "op": "delete_records", "family": "conformity_changes",
                   "keys": ["C-0031"]},
                  {"finding": "monitoring_limits:limits_not_from_sealed_baseline",
                   "op": "set_check_param", "check": "monitoring_limits", "param": "k",
                   "value": 5.0}])


def _check_inspection_report() -> None:
    nothing = inspection_report(audit_repair(PACK, SUITE, []))
    assert set(nothing) == {"system_id", "as_of", "checks", "green_before", "green_after",
                            "findings", "counts", "collateral", "evidence",
                            "suite_unchanged", "unsupported", "verdict"}, (
        f"the report has keys {sorted(nothing)} — return exactly the twelve documented")
    assert nothing["verdict"] == "rejected: findings unresolved", (
        f"doing nothing at all scored {nothing['verdict']!r}")
    assert nothing["green_before"] == 0 and nothing["checks"] == 6, (
        f"green_before {nothing['green_before']}, checks {nothing['checks']} — "
        "green_before is len(before_run['green']) and 'checks' is how many checks that run "
        "covered, not how many findings there are")
    assert [r["finding_id"] for r in nothing["findings"]] == [
        f["finding_id"] for f in inspect_pack(run_suite(PACK, SUITE))], (
        "the report's rows must stay in the findings' own order, worst first")
    silent = inspection_report(audit_repair(PACK, SUITE, SILENT_PLAN))
    assert silent["green_after"] == 6, (
        f"the silencing plan left {silent['green_after']} checks green, expected all six — "
        "that is the point of it")
    assert silent["counts"]["silenced"] == 6, (
        f"six checks green, {silent['counts']} — every one of those six went green without "
        "anything about the system changing")
    assert silent["verdict"] == "rejected: evidence silenced", (
        f"a pack with six green checks and six silenced findings scored "
        f"{silent['verdict']!r}. Silenced outranks unresolved: a pack that is honestly red "
        "is in better shape than one that has been made green")
    honest = inspection_report(
        audit_repair(PACK, SUITE, plan_repairs(PACK, inspect_pack(run_suite(PACK, SUITE)))))
    assert honest["verdict"] == "accepted" and honest["green_after"] == 6, (
        f"the honest repair scored {honest['verdict']!r}")
    assert honest["unsupported"] == [] and honest["suite_unchanged"] is True, (
        f"the honest repair reported unsupported={honest['unsupported']} and "
        f"suite_unchanged={honest['suite_unchanged']} — pass delta['totals'] and "
        "suite_change['unchanged'] through: every number it filed recomputes, and it "
        "edited no check")
    print(f"nothing done: {nothing['verdict']} · silenced: {silent['verdict']} · "
          f"repaired: {honest['verdict']}")


_try("exercise 8", _check_inspection_report, needs=_UPSTREAM["exercise 8"])
```

## 16. Common mistakes

- **Treating a check that did not run as a check that passed.** It is the cheapest silencing
  there is and it is usually nobody's idea: a check errors in CI, somebody marks it skipped,
  and the pack is green a sprint later. `run_suite` has three statuses for this reason, and
  `verdict` is `"pass"` only when nothing is red *and* nothing is unverified.
- **Auditing the repair ledger instead of the pack.** The ledger is the repairer's account
  of what they did. `unsupported_values` recomputes from the retained evidence and never
  reads it, which is why a lying journal entry — a source named for a number that does not
  come back — is caught anyway.
- **Reading `conserved` as "honest".** It means one thing: nothing was deleted. The
  re-chained log removes nothing and rewrites every entry after the hole, which the table
  in section 12 counts for you.
- **Flagging every change to the suite.** Tightening a threshold is not silencing, and an
  audit that pages on both is muted within a month — which is module 7's lesson arriving
  from a different direction.
- **Stopping halfway.** Voiding the four hundred implausible reviews without withholding the
  decisions they support leaves four hundred actions standing on nothing. The check goes red
  with a *different* reason code, the finding is `unfixed`, and unfixed is an honest place
  to be.
- **Fixing a finding with a value you typed.** A metric can be declared only from something
  that was measured. The evaluation split is retained in the pack precisely so the interval
  can be recomputed by whoever reads it — including by you, ten minutes after you wrote it.
- **Restoring the wrong cold-store segment.** One of the two exports holds entries the log
  already has. Restoring it succeeds, changes nothing, and leaves the check red; match the
  segment's `covers` range against the missing sequence numbers instead of taking the first.
- **Judging a repair by the pipeline.** Four of the five repairs in `TEMPTING_PLANS` leave
  a check the pipeline calls green, and the fifth switches the check off — which a
  two-status runner reports as a pass. Five of the six findings in `SILENT_PLAN` are green
  *and* removed, rewrote or hid the evidence the finding was about.

```
def _show_half_done() -> None:
    """Void the implausible reviews, stop there, and read what the check says next."""
    findings = inspect_pack(run_suite(PACK, SUITE))
    finding = next(f for f in findings if f["check"] == "oversight_plausibility")
    half = [{"finding": finding["finding_id"], "op": "void_verifications",
             "ids": [k.split(":", 1)[1] for k in finding["evidence_keys"]
                     if k.startswith("verification:")],
             "authority": "u:sanne",
             "reason": "recorded faster than a case can be read"}]
    result = audit_repair(PACK, SUITE, half)
    before = result["before_run"]["results"]["oversight_plausibility"]
    after = result["after_run"]["results"]["oversight_plausibility"]
    verdict = result["audit"]["verdicts"][finding["finding_id"]]
    print(f"  before  {before['status']:6s} {before['reason_code']:38s} "
          f"{before['affected']}/{before['population']}")
    print(f"  after   {after['status']:6s} {after['reason_code']:38s} "
          f"{after['affected']}/{after['population']}")
    print(f"  audit   {verdict['verdict']}\n")
    print("Voiding the reviews moved the same decisions from one red to another: they are no\n"
          "longer verified by somebody who cannot have read them, they are not verified at\n"
          "all, and the actions still stand. The audit calls that unfixed, and unfixed is a\n"
          "better place to be than green.")


_try("half a repair", _show_half_done, needs=_PIPELINE)
```

## 17. Self-check

1. A colleague deletes the four hundred implausible verification records and the decisions
   they supported. Every check is green. Your audit should report:
   - (a) repaired — the check is green and no threshold moved
   - (b) silenced — the evidence the finding was about is no longer in the pack, which is
         not the same as the system having changed
   - (c) unfixed — deletion never fixes anything
   - (d) repaired, but with a note in the report

2. `evidence_delta(...)["conserved"]` comes back `True` for a repair. That entitles you to
   say:
   - (a) the repair was honest
   - (b) the pack is unchanged
   - (c) nothing was deleted — and nothing else: a repair that rewrote every digest in the
         pack and removed nothing is also "conserved"
   - (d) every check is green

3. A repair lowers `max_reviews_per_hour` from 60 to 20. `suite_delta` should put it in:
   - (a) "tightened", and `unchanged` stays True — the check got stricter, and flagging it
         teaches people to ignore the audit
   - (b) "loosened", because the parameter moved
   - (c) "unknown", because thresholds are judgement calls
   - (d) "disabled"

4. A repair voids the implausible verifications but does not withhold the decisions that
   rested on them. The oversight check comes back red with the reason code
   `standing_action_without_verification`. The finding's verdict is:
   - (a) silenced — the reason code changed
   - (b) repaired — the implausible reviews are gone from the standing set
   - (c) silenced — voiding is a deletion with extra steps
   - (d) unfixed — the check is red, the pack says so, and a half-done repair that reports
         itself as half-done is the best of the three outcomes here

5. Why does `unsupported_values` never read `pack["repair_ledger"]`?

## What you built, and what the programme leaves you with

An inspection that runs the nine modules' tools over somebody else's pack and reports a
check nobody ran as the worst finding on the page; a repair expressed as operations against
the evidence rather than as edits to a file; and the audit that decides whether the repair
was a repair — three independent measurements, two of which never trust a word the repairer
wrote down.

That audit is the only thing in this programme that cannot be borrowed from a template. A
conformity pack is a promise that somebody will check, and the whole apparatus — the hash
chain in module 1, the typed sources in module 3, the retained evaluation split in module 6,
the baseline sealed before the period it judges in module 7, the assessment ledger in module
8 — exists for one purpose: so that a year from now, somebody who was not there can tell a
number that was measured from a number that was needed.
